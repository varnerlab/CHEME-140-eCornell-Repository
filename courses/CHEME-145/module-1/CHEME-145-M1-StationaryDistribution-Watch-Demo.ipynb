# Stationary Distribution
The stationary distribution describes the long-run equilibrium behavior of a Markov chain. After running the chain for many steps, the probability of being in each state stabilizes, regardless of the initial state. This equilibrium distribution is fundamental to understanding Markov chain behavior, as it characterizes where the system spends its time in the limit.

> __Definition:__ The **stationary distribution** of a Markov chain is a probability vector $\bar{\pi}$ that remains unchanged under the transition dynamics: $\bar{\pi} = \bar{\pi} \mathbf{P}$. If the system starts in distribution $\bar{\pi}$, it remains in $\bar{\pi}$ indefinitely. Moreover, $\bar{\pi}_i$ represents the long-run fraction of time the chain spends in state $i$.
>
> __Existence and Uniqueness:__ A unique stationary distribution $\bar{\pi}$ exists if the Markov chain is **ergodic**, meaning (1) **Irreducible**: every state is reachable from every other state, and (2) **Aperiodic**: the chain does not oscillate in a fixed pattern. For such chains, starting from any initial distribution $\pi_{0}$, the distribution converges to $\bar{\pi}$ as $k \to \infty$: $\pi_k = \pi_0 \mathbf{P}^k \to \bar{\pi}$. The powers of the transition matrix converge to a limiting matrix where each row equals $\bar{\pi}$, reflecting that the long-run distribution is independent of the starting state.

## Computing the Stationary Distribution

### Power Iteration Method
We can estimate $\bar{\pi}$ by iteratively applying the transition matrix starting from any initial distribution $\pi_0$:

$$
\begin{align*}
\pi_1 &= \pi_0 \mathbf{P} \\
\pi_2 &= \pi_1 \mathbf{P} = \pi_0 \mathbf{P}^2 \\
\pi_3 &= \pi_2 \mathbf{P} = \pi_0 \mathbf{P}^3 \\
&\vdots \\
\pi_k &= \pi_0 \mathbf{P}^k
\end{align*}
$$

#### Algorithm
Let's develop an iterative algorithm for estimating the stationary distribution $\bar{\pi}$.

__Initialize__: Given an ergodic transition matrix $\mathbf{P}$ with $N$ states, tolerance parameter $\epsilon$, and maximum number of iterations $T$. Initialize the iteration counter $t\gets 0$, the uniform initial distribution $\pi_{\circ} \gets \left[1/N,\ldots,1/N\right]$, and $\texttt{converged}\gets\texttt{false}$.

While $\texttt{converged}$ is $\texttt{false}$ __do__:
1. Compute: $\pi^{\prime} \gets \pi_{\circ}\mathbf{P}$.
2. Check for convergence:
    - If $\lVert \pi^{\prime} - \pi_{\circ} \rVert_{1} \leq \epsilon$, then set $\texttt{converged}\gets\texttt{true}$ and $\bar{\pi}\gets\pi^{\prime}$.
    - If $\lVert \pi^{\prime} - \pi_{\circ} \rVert_{1} > \epsilon$, update $t\gets{t+1}$ and $\pi_{\circ}\gets\pi^{\prime}$.
3. Update the $\texttt{converged}$ flag:
    - If $t\geq{T}$, then set $\texttt{converged}\gets\texttt{true}$ and $\bar{\pi}\gets\pi^{\prime}$.

__Output__: The stationary distribution $\bar{\pi}$.

The convergence rate depends on the spectral gap (difference between the largest and second-largest eigenvalue magnitudes of $\mathbf{P}$), with larger gaps producing faster convergence.

### Alternative: Eigenvalue Approach
The stationary distribution $\bar{\pi}$ is the left eigenvector of $\mathbf{P}$ corresponding to eigenvalue $\lambda = 1$, normalized to sum to 1. This can be computed by solving $\bar{\pi}\mathbf{P} = \bar{\pi}$ subject to $\sum_i \bar{\pi}_i = 1$, which is equivalent to finding the null space of $(\mathbf{P}^{\top} - \mathbf{I})$ for the unique eigenvector with eigenvalue 1.
