# Hidden Markov Model (HMM)
A Hidden Markov Model extends the standard Markov chain by introducing a layer of indirection: the true state sequence is hidden, and we only observe noisy or indirect signals emitted by those hidden states. This two-layer structure makes HMMs powerful for modeling systems where we cannot directly observe the underlying process but can see its effects.

> __Definition:__ A Hidden Markov Model is defined by the tuple $\mathcal{H} = (\mathcal{S}, \mathcal{O}, \mathbf{P}, \mathbf{E}, \pi_0)$, where:
> * **Hidden state space** $\mathcal{S}$: The set of unobservable states that the system occupies
> * **Observable output space** $\mathcal{O}$: The set of signals or emissions we can actually measure
> * **Transition matrix** $\mathbf{P}$: Governs how the hidden states evolve over time, where $p_{ij} = P(s_{t+1} = j \mid s_t = i)$
> * **Emission matrix** $\mathbf{E}$: Controls the probability of observing output $o$ given hidden state $s$, where $e_{ij} = P(o_t = j \mid s_t = i)$
> * **Initial distribution** $\pi_0$: The probability distribution over hidden states at time $t=0$
>
> The hidden states evolve according to $\mathbf{P}$ (a standard Markov chain), while at each time step, the current hidden state probabilistically generates an observable output according to $\mathbf{E}$.

## Why "Hidden"?
In a standard Markov chain, we directly observe which state the system occupies at each time step. In an HMM, the state sequence $s_1, s_2, \ldots$ is latent (hidden), and we only observe emissions $o_1, o_2, \ldots$ that provide indirect evidence about the hidden states. This structure naturally models scenarios where:
* **Speech recognition**: Hidden states are phonemes, observations are acoustic signals
* **Biological sequence analysis**: Hidden states are gene structure elements (exon, intron), observations are DNA nucleotides
* **Financial markets**: Hidden states are market regimes (bull, bear), observations are price movements
* **Sentiment analysis**: Hidden states are emotional states, observations are words in text

## Key Properties
The emission matrix $\mathbf{E}$ introduces uncertainty between hidden states and observations. Each row $i$ of $\mathbf{E}$ must sum to 1, representing the distribution over possible observations when in hidden state $i$. The degree of noise in the observations depends on how "peaked" these distributions are—high emission probability for the correct observation means low noise, while uniform emission probabilities mean maximum uncertainty.

## Fundamental Problems
Three classical problems drive HMM applications:

1. **Evaluation**: Given an observation sequence $\mathbf{o} = (o_1, \ldots, o_T)$ and model parameters $(\mathbf{P}, \mathbf{E}, \pi_0)$, compute the probability $P(\mathbf{o} \mid \mathcal{H})$. Solved via the forward algorithm.

2. **Decoding**: Given observations $\mathbf{o}$ and model $\mathcal{H}$, find the most likely hidden state sequence $\mathbf{s}^* = \arg\max_{\mathbf{s}} P(\mathbf{s} \mid \mathbf{o}, \mathcal{H})$. Solved via the Viterbi algorithm.

3. **Learning**: Given only observations $\mathbf{o}$, estimate the model parameters $(\mathbf{P}, \mathbf{E}, \pi_0)$ that maximize $P(\mathbf{o} \mid \mathcal{H})$. Solved via the Baum-Welch (EM) algorithm.

These algorithms leverage dynamic programming to efficiently compute quantities that would be intractable by brute-force enumeration over all possible state sequences.
