# Classical Decision Systems: Preferences, Utility, Risk, and Choice
This module introduces the classical theory of rational decision-making: how an agent ranks alternatives, represents that ranking with a utility function, allocates a scarce budget across goods, and chooses under uncertainty. These tools are the foundation for the sequential and probabilistic decision models in the rest of the course: Markov models, hidden Markov models, and Markov decision processes.

> __Learning Objectives.__
>
> By the end of this module, you will be able to define and demonstrate mastery of the following key concepts:

> * __Preferences and utility functions:__ State the rationality axioms (completeness and transitivity) and explain how a preference relation over a set of alternatives is represented by an ordinal utility function $U:\mathcal{X}\rightarrow\mathbb{R}$ that is unique up to an order-preserving transformation.
> * __Constrained choice and expected utility:__ Solve a constrained utility-maximization problem with the Lagrange multiplier method, and extend deterministic choice to choice under risk using the expected utility hypothesis, the certainty equivalent, and the Arrow–Pratt measures of risk aversion.
> * __Discrete choice models:__ Decompose utility into observed and unobserved components with a random utility model $U_{ij}=V_{ij}+\varepsilon_{ij}$, and derive the multinomial logit choice probabilities of McFadden under the assumption of independent and identically distributed Gumbel errors.

Each section adds one ingredient: first a ranking, then a budget, then uncertainty, and finally unobserved features. Let's get started!

## Preferences and Rational Choice
We model a decision-maker as an __agent__ that selects from a set of alternatives. Let $\mathcal{X}=\{x_1,x_2,\ldots,x_n\}$ be a finite set of alternatives, where each alternative is an object, a bundle of goods, or a state of the world. The agent compares alternatives with a __preference relation__ $\succeq$, where $x_i \succeq x_j$ reads "the agent likes $x_i$ at least as much as $x_j$." From $\succeq$ we define two derived relations:
* __Strict preference:__ $x_i \succ x_j$ if $x_i \succeq x_j$ and not $x_j \succeq x_i$.
* __Indifference:__ $x_i \sim x_j$ if $x_i \succeq x_j$ and $x_j \succeq x_i$.

A preference relation is __rational__ if it satisfies two axioms:
1. __Completeness:__ for every pair $x_i, x_j \in \mathcal{X}$, $x_i \succeq x_j$ or $x_j \succeq x_i$ (or both). The agent can compare any two alternatives.
2. __Transitivity:__ for every $x_i, x_j, x_k \in \mathcal{X}$, if $x_i \succeq x_j$ and $x_j \succeq x_k$, then $x_i \succeq x_k$. The agent's comparisons are consistent.

These axioms do not tell the agent what to prefer; they require only that the preferences be complete and consistent. An agent whose preferences violate transitivity can be led through a cycle of trades that returns to the starting bundle with less wealth, which is why transitivity is treated as a minimum condition for rationality.
___


## Utility Functions
When the alternative set is finite and preferences are rational, the ranking is represented by a __utility function__. A utility function $U:\mathcal{X}\rightarrow\mathbb{R}$ assigns a real number, in units of _utils_, to each alternative such that:
$$
U(x_i) \geq U(x_j) \quad\Longleftrightarrow\quad x_i \succeq x_j.
$$
The utility function carries the ordering of alternatives, and nothing more:
* __Ordinal, not cardinal:__ utility rank-orders alternatives but does not measure the size of the difference between them. If $U(x_i)=10$ and $U(x_j)=5$, then $x_i$ is preferred to $x_j$; it does not follow that $x_i$ is "twice as good."
* __Unique up to an order-preserving transformation:__ if $U$ represents $\succeq$ and $f:\mathbb{R}\rightarrow\mathbb{R}$ is strictly increasing, then $f\!\left(U\right)$ represents the same preferences.

### Common utility functions
Let $\mathbf{x}=(x_1,\ldots,x_n)$ with $x_i \geq 0$ denote a bundle of $n$ goods, and let $\boldsymbol{\alpha}=(\alpha_1,\ldots,\alpha_n)$ with $\alpha_i \geq 0$ denote parameters. Four forms appear throughout the course:

| Utility function | Form | Models |
| :--- | :--- | :--- |
| Linear | $U(\mathbf{x}) = \boldsymbol{\alpha}^{\top}\mathbf{x} = \sum_{i=1}^{n}\alpha_i x_i$ | goods that are perfect substitutes |
| Logarithmic | $U(\mathbf{x}) = \ln\!\left(\boldsymbol{\alpha}^{\top}\mathbf{x} + \beta\right),\ \beta>0$ | a single good with diminishing returns |
| Cobb–Douglas | $U(\mathbf{x}) = \prod_{i=1}^{n} x_i^{\alpha_i},\ \sum_{i=1}^{n}\alpha_i = 1$ | goods consumed together |
| Leontief | $U(\mathbf{x}) = \min\!\left(\dfrac{x_1}{\alpha_1},\ldots,\dfrac{x_n}{\alpha_n}\right),\ \alpha_i>0$ | perfect complements (fixed proportions) |

### Marginal utility
The __marginal utility__ of good $i$ is the partial derivative of $U$ with respect to $x_i$:
$$
\bar{U}_{i} \equiv \frac{\partial U}{\partial x_i},
$$
the change in utility from a small change in the quantity of good $i$. Many utility functions have positive marginal utility ("more is better," $\bar{U}_i > 0$) that decreases as $x_i$ grows (diminishing marginal utility). Marginal utility is the link between the ranking and the constrained choice problem in the next section.

## Constrained Choice: Budgets and the Lagrangian
A ranking is not enough; a real agent faces __constraints__. Suppose the agent has a budget of $I>0$ units of a scarce resource (money, time) and good $i$ has unit cost $c_i>0$. The agent maximizes utility subject to the budget:
$$
\begin{aligned}
\max_{\mathbf{x}} \quad & U(x_1,\ldots,x_n) \\
\text{subject to}\quad & \sum_{i=1}^{n} c_i x_i = I, \\
& x_i \geq 0, \quad i=1,\ldots,n.
\end{aligned}
$$
The constraint $\sum_i c_i x_i = I$ is the __budget constraint__. It restricts the agent to bundles that are affordable, replacing "more is always better" with "more of one good means less of another."

### The Lagrange multiplier method
To maximize $U(\mathbf{x})$ subject to the equality constraint, form the __Lagrangian__
$$
\mathcal{L}(\mathbf{x},\lambda) = U(\mathbf{x}) + \lambda\left(I - \sum_{i=1}^{n} c_i x_i\right),
$$
where $\lambda$ is the __Lagrange multiplier__ for the budget constraint. At an interior optimum the partial derivatives of $\mathcal{L}$ vanish, giving the first-order optimality conditions:
$$
\begin{aligned}
\frac{\partial \mathcal{L}}{\partial x_i} &= \bar{U}_i - \lambda\, c_i = 0, \quad i=1,\ldots,n, \\
\frac{\partial \mathcal{L}}{\partial \lambda} &= I - \sum_{i=1}^{n} c_i x_i = 0.
\end{aligned}
$$
This is a system of $n+1$ equations in the $n+1$ unknowns $(x_1,\ldots,x_n,\lambda)$. The first set rearranges to
$$
\frac{\bar{U}_i}{c_i} = \lambda \quad \text{for all } i,
$$
the rule that __marginal utility per unit cost is equal across all goods__ at the optimum. The common value $\lambda$ is the marginal utility of an extra unit of budget, the shadow price of the constraint.

### Marginal rate of substitution
Dividing the first-order condition for good $i$ by that for good $j$ removes $\lambda$:
$$
\underbrace{\frac{\bar{U}_i}{\bar{U}_j}}_{\text{MRS}_{ij}} = \frac{c_i}{c_j}.
$$
The left side is the __marginal rate of substitution__ (MRS): the rate at which the agent trades good $j$ for good $i$ while holding utility constant. At the optimum the MRS equals the price ratio. Geometrically, the indifference curve is tangent to the budget line at the optimal bundle.
___


## Decision Under Risk: Expected Utility
So far each choice yields a certain outcome. Now suppose the outcome is uncertain. A __lottery__ is a set of outcomes $\{w_1,\ldots,w_m\}$ with probabilities $\{p_1,\ldots,p_m\}$, where $p_k \geq 0$ and $\sum_{k=1}^{m} p_k = 1$. Here $w_k$ denotes a level of wealth (the payoff) in outcome $k$.

The __expected value__ of the payoff is $\mathbb{E}[W]=\sum_{k=1}^{m} p_k\, w_k$. A first idea is to rank lotteries by expected value, but this fails. The __St. Petersburg paradox__ (Bernoulli, 1738) is a game that pays $2^{k}$ dollars if the first head appears on toss $k$, which has probability $2^{-k}$. Its expected payoff is
$$
\mathbb{E}[W] = \sum_{k=1}^{\infty} 2^{-k}\cdot 2^{k} = \sum_{k=1}^{\infty} 1 = \infty,
$$
yet few people would pay more than a small amount to play. Ranking by expected value alone does not match observed choices.

The __expected utility hypothesis__ of von Neumann and Morgenstern resolves this: a rational agent ranks lotteries by the expected utility of wealth, not by expected wealth:
$$
\mathbb{E}[U(W)] = \sum_{k=1}^{m} p_k\, U(w_k).
$$
When the von Neumann–Morgenstern axioms (completeness, transitivity, continuity, and independence) hold, preferences over lotteries are represented by $\mathbb{E}[U(W)]$, and the agent chooses the action $a$ from a set $\mathcal{A}$ that maximizes expected utility:
$$
a^{\star} = \arg\max_{a\in\mathcal{A}} \; \mathbb{E}[\,U(W \mid a)\,].
$$
This principle of maximum expected utility is the objective that the Markov decision processes in Module 4 maximize over sequences of actions.

## Risk Aversion, Certainty Equivalent, and Risk Premium
The shape of the utility function $U$ encodes the agent's attitude toward risk. Compare a risky lottery $W$ with its expected value $\mathbb{E}[W]$ received for certain. By Jensen's inequality, the curvature of $U$ determines the ranking:

| Curvature of $U$ | Inequality | Risk attitude |
| :--- | :--- | :--- |
| Concave ($U''<0$) | $\mathbb{E}[U(W)] \leq U(\mathbb{E}[W])$ | risk averse |
| Linear ($U''=0$) | $\mathbb{E}[U(W)] = U(\mathbb{E}[W])$ | risk neutral |
| Convex ($U''>0$) | $\mathbb{E}[U(W)] \geq U(\mathbb{E}[W])$ | risk seeking |

where $U'(w)$ and $U''(w)$ are the first and second derivatives of $U$ with respect to wealth $w$.

### Certainty equivalent and risk premium
The __certainty equivalent__ $\mathrm{CE}$ is the guaranteed wealth that gives the same utility as the risky lottery:
$$
U(\mathrm{CE}) = \mathbb{E}[U(W)] \quad\Longrightarrow\quad \mathrm{CE} = U^{-1}\!\left(\mathbb{E}[U(W)]\right).
$$
The agent is indifferent between receiving $\mathrm{CE}$ for certain and holding the lottery. The __risk premium__ is the gap between the expected wealth and the certainty equivalent:
$$
\pi = \mathbb{E}[W] - \mathrm{CE}.
$$
For a risk-averse agent $U$ is concave, so $\mathrm{CE} \leq \mathbb{E}[W]$ and $\pi \geq 0$: the agent will give up $\pi$ units of expected wealth to remove the risk. This $\pi$ is the most the agent will pay for full insurance against the lottery.

### Arrow–Pratt measures of risk aversion
The curvature of $U$ is summarized by the __Arrow–Pratt coefficient of absolute risk aversion__:
$$
A(w) = -\frac{U''(w)}{U'(w)},
$$
and the __coefficient of relative risk aversion__:
$$
R(w) = -\,w\,\frac{U''(w)}{U'(w)} = w\,A(w).
$$
For a risk-averse agent $U'(w)>0$ and $U''(w)<0$, so $A(w)>0$. A larger $A(w)$ means more curvature, a larger risk premium for a given gamble, and a smaller certainty equivalent.

## Discrete Choice Models
The decisions above assume the agent knows utility exactly. In practice the modeler observes only some of the features that drive a choice. __Discrete choice models__ split utility into an observed component and an unobserved component.

### Random utility model
For individual $i$ choosing among $J$ alternatives, the utility of alternative $j$ is
$$
U_{ij} = V_{ij} + \varepsilon_{ij},
$$
where $V_{ij}$ is the __deterministic__ component built from observed features, for example $V_{ij}=\boldsymbol{\beta}^{\top}\mathbf{x}_{ij}$ with feature vector $\mathbf{x}_{ij}$ and weights $\boldsymbol{\beta}$, and $\varepsilon_{ij}$ is the __random__ component capturing unobserved features. The individual chooses the alternative with the highest utility:
$$
i \text{ chooses } j \quad\Longleftrightarrow\quad U_{ij} \geq U_{ik} \;\; \text{for all } k=1,\ldots,J.
$$
Because $\varepsilon_{ij}$ is random, the choice is described by a probability rather than a single outcome.

### The logit model
McFadden showed that if the errors $\varepsilon_{ij}$ are independent and identically distributed (IID) across the $J$ alternatives with a Gumbel (Type-I extreme value) distribution, then the probability that individual $i$ chooses alternative $j$ has the closed form
$$
P_{ij} = \frac{\exp\!\left(\mu V_{ij}\right)}{\sum_{k=1}^{J}\exp\!\left(\mu V_{ik}\right)},
$$
the __multinomial (conditional) logit__ model, where $\mu>0$ is a scale parameter, often normalized to $\mu=1$. With two alternatives this reduces to the binary logit, and the choice is a Bernoulli random variable: alternative $j$ is selected with probability $P_{ij}$ and not selected with probability $1-P_{ij}$.

### Independence of irrelevant alternatives
In the logit model the ratio of the choice probabilities for two alternatives depends only on those two alternatives:
$$
\frac{P_{ij}}{P_{ik}} = \frac{\exp\!\left(\mu V_{ij}\right)}{\exp\!\left(\mu V_{ik}\right)} = \exp\!\left(\mu\,(V_{ij}-V_{ik})\right),
$$
independent of the other alternatives in the choice set. This is the __independence of irrelevant alternatives (IIA)__ property. It makes the model simple to estimate, but it can be unrealistic when alternatives are close substitutes.

## Summary
This module developed the classical pipeline of rational decision-making: rank alternatives with a preference relation, represent the ranking with a utility function, maximize utility subject to a budget constraint, and extend the framework to choice under risk and to choice with unobserved features.

> __Key takeaways:__
>
> 1. **Preferences, utility, and constrained choice:** A rational agent has complete and transitive preferences, which an ordinal utility function $U:\mathcal{X}\rightarrow\mathbb{R}$ represents up to an order-preserving transformation. Subject to a budget constraint, the optimal bundle equalizes marginal utility per unit cost across goods, $\bar{U}_i/c_i=\lambda$, found with the Lagrange multiplier method.
> 2. **Expected utility and risk aversion:** Under the von Neumann–Morgenstern axioms, an agent ranks risky lotteries by expected utility $\mathbb{E}[U(W)]$ rather than by expected wealth. The curvature of $U$ sets the risk attitude: a concave $U$ implies risk aversion, a positive risk premium $\pi=\mathbb{E}[W]-\mathrm{CE}$, and a positive Arrow–Pratt coefficient $A(w)=-U''(w)/U'(w)$.
> 3. **Discrete choice and the logit model:** A random utility model $U_{ij}=V_{ij}+\varepsilon_{ij}$ separates observed from unobserved utility. With IID Gumbel errors, the choice probabilities take the multinomial logit form $P_{ij}=\exp(\mu V_{ij})/\sum_{k}\exp(\mu V_{ik})$, which exhibits the independence of irrelevant alternatives property.

Classical decision theory turns a verbal description of what an agent wants into a computable objective that can be maximized, estimated from data, and embedded in the sequential decision models developed in the rest of the course. The maximum expected utility principle introduced here is the same objective that Markov decision processes optimize over sequences of actions in Module 4.
