# Foundations of Markov Models and Markov Chains
This module introduces Markov models, a foundational tool for modeling systems that evolve randomly over time. Markov models appear across many domains, from optimizing operations and pricing assets in finance to decoding biological signals and training machine-learning algorithms.

> __Learning Objectives:__
> 
> By the end of this module, you will be able to define the following key concepts:
>
> * __The Markov Property:__ A random process is said to follow the _Markov property_ when its next state depends only on the present state, not on the sequence of states that preceded it. The future is conditionally independent of the past, given the present.
> * __Visible vs. Hidden Chains:__ A _visible_ (or plain) Markov chain models a sequence of directly observed states, where each state depends solely on the previous one. A _hidden Markov model_ (HMM) is a Markov chain whose true state is hidden; it emits _observable signals_ governed by state-dependent probabilities.
> * __Stationary Distribution:__ The stationary distribution describes the chain’s long-run equilibrium behavior. It is a probability vector $\bar{\pi}$ satisfying $\bar{\pi} = \bar{\pi}\mathbf{P}$ (with $\sum_i \bar{\pi}_i = 1$), meaning that a chain started in $\bar{\pi}$ remains in $\bar{\pi}$. Equivalently, $\bar{\pi}^\top$ is an eigenvector of $\mathbf{P}^\top$ with eigenvalue 1 (a left eigenvector of $\mathbf{P}$).

Though the Markov assumption may seem restrictive, it often captures real-world dynamics well: stock returns, for instance, depend more on today’s and yesterday’s prices than on the entire trading history, and a cell’s next state may depend more on its current condition than on its full lineage.
