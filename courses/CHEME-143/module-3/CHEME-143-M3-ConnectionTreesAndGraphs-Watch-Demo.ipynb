# Special Graph Types
While general graphs provide a flexible framework for modeling relationships, certain specialized graph structures possess unique properties that make them particularly useful for specific applications. 

We've already seen one of these special graphs: __trees__.

## Trees
A __tree__ is a connected acyclic undirected graph. Trees have $|\mathcal{V}| - 1$ edges and possess a unique path between any two vertices. Key properties of trees include:
* __Acyclic__: Contains no cycles or loops
* __Connected__: Every vertex is reachable from every other vertex
* __Minimal connectivity__: Removing any edge disconnects the graph
* __Maximal acyclicity__: Adding any edge creates exactly one cycle

Trees are fundamental in computer science, appearing in data structures (binary search trees, heaps), decision models, and hierarchical organization systems.

Another important class of graphs is __complete graphs__.

## Complete Graphs
A __complete graph__ $K_{n}$ is a simple undirected graph where every pair of distinct vertices is connected by a unique edge. For a complete graph with $n$ vertices:
* Number of edges: $|\mathcal{E}| = \frac{n(n-1)}{2}$
* Every vertex has degree $n-1$
* Graph density $\rho = 1$ (maximally dense)

Complete graphs represent scenarios where every entity is directly connected to every other entity, such as fully-connected communication networks or round-robin tournament structures.

## Bipartite Graphs
A __bipartite graph__ $\mathcal{G} = (\mathcal{V}_{1} \cup \mathcal{V}_{2}, \mathcal{E})$ is a graph whose vertex set can be partitioned into two disjoint sets $\mathcal{V}_{1}$ and $\mathcal{V}_{2}$ such that every edge connects a vertex in $\mathcal{V}_{1}$ to a vertex in $\mathcal{V}_{2}$. No edges exist within either partition.

#### Mathematical Properties of Bipartite Graphs
* __2-colorable__: Vertices can be colored with exactly two colors such that no adjacent vertices share the same color
* __No odd cycles__: A graph is bipartite if and only if it contains no odd-length cycles
* __Partite sets__: The two vertex sets $\mathcal{V}_{1}$ and $\mathcal{V}_{2}$ are called partite sets or parts

#### Complete Bipartite Graphs
A __complete bipartite graph__ $K_{m,n}$ is a bipartite graph where every vertex in $\mathcal{V}_{1}$ (with $|\mathcal{V}_{1}| = m$) is connected to every vertex in $\mathcal{V}_{2}$ (with $|\mathcal{V}_{2}| = n$). Properties include:
* Total edges: $|\mathcal{E}| = m \times n$
* Degrees: Every vertex in $\mathcal{V}_{1}$ has degree $n$, every vertex in $\mathcal{V}_{2}$ has degree $m$

#### Applications of Bipartite Graphs
Bipartite graphs naturally model relationships between two distinct types of entities:

* __Matching Problems__: Assigning workers to jobs, students to courses, or participants to activities
* __Recommendation Systems__: Users connected to items they've interacted with (movies, products, articles)
* __Biological Networks__: Gene-disease associations, protein-drug interactions, or metabolic pathways connecting metabolites to reactions
* __Resource Allocation__: Machines assigned to tasks, servers handling client requests, or channels allocated to users
* __Collaboration Networks__: Authors connected to papers, actors connected to movies, or researchers connected to projects

#### Detecting Bipartiteness
A graph is bipartite if and only if it can be 2-colored. This can be verified using breadth-first search (BFS) or depth-first search (DFS):
1. Start with any vertex and assign it color 1
2. Assign all its neighbors color 2
3. Continue coloring neighbors with alternating colors
4. If a conflict arises (adjacent vertices would have the same color), the graph is not bipartite

#### Storage Considerations for Bipartite Graphs
For bipartite graphs with partite sets of significantly different sizes ($|\mathcal{V}_{1}| \gg |\mathcal{V}_{2}|$ or vice versa), specialized representations can be more efficient:
* __Biadjacency Matrix__: An $|\mathcal{V}_{1}| \times |\mathcal{V}_{2}|$ matrix rather than a full $|\mathcal{V}| \times |\mathcal{V}|$ adjacency matrix
* __Partitioned Adjacency Lists__: Separate adjacency lists for each partite set, reducing memory overhead
* __Incidence Representation__: Directly represent the bipartite structure as mappings between the two vertex types

These specialized representations are particularly valuable when $|\mathcal{V}_{1}| + |\mathcal{V}_{2}| \ll |\mathcal{V}_{1}| \times |\mathcal{V}_{2}|$, providing significant space savings over standard graph representations.
