## Memoization: Conceptual Overview
Memoization is an optimization technique to speed up recursive algorithms by storing the results of expensive function calls and reusing them when the same inputs occur again. 

> __Key idea__: The critical idea of memoization is to do only what is necessary! Instead of recomputing things, why not try to recycle your expensive calculations if possible. If we've already computed a value, store it in case we need it again.

Let's think about a classic example: calculating Fibonacci numbers. The naive recursive approach has an exponential time complexity because it recalculates the same Fibonacci numbers multiple times.

The diagram shows the recursive call tree for computing `fibonacci(4)` using the naive recursive approach. Here's what's happening:

**The Problem:** To calculate `fibonacci(4)`, the algorithm needs to compute `fibonacci(3)` and `fibonacci(2)`. But to compute `fibonacci(3)`, it needs `fibonacci(2)` and `fibonacci(1)`. Notice that `fibonacci(2)` gets computed **twice** - once for the `fibonacci(4) → fibonacci(3)` branch and once for the `fibonacci(4) → fibonacci(2)` branch.

**The Inefficiency:** Even in this small example, we can see redundant calculations:
- `fibonacci(2)` is computed 2 times
- `fibonacci(1)` is computed 3 times  
- `fibonacci(0)` is computed 2 times

The highlighted region shows values that could be **cached (memoized)** after their first calculation. Instead of recomputing `fibonacci(2)` the second time, we could just look it up from our stored results!

**The Exponential Growth:** As `n` increases, this redundancy explodes exponentially with $\mathcal{O}(2^n)$ time complexity. The naive recursive approach makes an exponentially growing number of function calls, but with memoization, it only needs $\mathcal{O}(n)$ calls, one for each unique value from 0 to n.
