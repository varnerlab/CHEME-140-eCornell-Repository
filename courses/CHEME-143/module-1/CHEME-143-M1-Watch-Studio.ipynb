# Introduction to Recursion
Recursion is a programming technique where a function calls itself to solve smaller instances of the same problem. It is often used to solve problems that can be broken down into smaller, similar subproblems. Recursion can be a powerful tool for solving complex problems in a more elegant and readable way. 

#### Motivation
* Recursion provides a direct mechanism for divide-and-conquer algorithms by having each function call split the problem into similarly structured subproblems, invoke itself to solve them, and then combine their results. This approach leads to concise, modular code that mirrors the mathematical recurrence of many efficient algorithms—such as Merge Sort and Quick Sort—while achieving their optimal time complexities.
* Recursion excels at expressing divide-and-conquer and other self-similar problems in concise, readable code that directly mirrors mathematical definitions. However, it can incur extra overhead from function calls and stack usage—leading to potential performance penalties or stack overflows on deep recursion—and may be harder to debug and reason about compared to equivalent iterative solutions.
