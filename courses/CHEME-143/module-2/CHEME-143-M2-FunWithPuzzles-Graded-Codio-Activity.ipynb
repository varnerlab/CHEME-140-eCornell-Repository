# Graded Codio Activity: Building Probabilistic Cellular Automata
In this activity, we will analyze a program that computes two-color one-dimensional cellular automata. In particular,we'll simulate [Rule 30](https://en.wikipedia.org/wiki/Rule_30), a one-dimensional cellular automaton rule introduced by Stephen Wolfram in 1983. It is particularly interesting because it produces complex, seemingly random patterns from simple, well-defined rules.

We'll look at two formulations of this model. After we setup the compuational environment, we'll simulate a system described by a deterministic one-dimensional cellular automaton in Task 1, and then in Task 2 we'll explore a probabilistic version of the same model.

This is a fascinating area of study that combines elements of physics, computer science, and mathematics. Cellular automata have been used to model a wide range of phenomena, from the behavior of gases to the spread of diseases. Very cool.

```julia
# what rule do we want to see?
colors = 2; # black and white
radius = 3; # radius of the rule - we look at three states
rule_index = 30; # what rule do we want to see?
number_of_steps = 240;
number_of_states = 2^8;
initial = zeros(Int64, number_of_states);
initial[round(number_of_states/2) |> Int] = 1;
```

## Task 1: Deterministic Cellular Automata
In this task, we will implement a deterministic one-dimensional cellular automaton [using Rule 30](https://en.wikipedia.org/wiki/Rule_30). We'll start with a simple initial condition and evolve the system over a number of time steps.


```julia
rulemodel = build(MyOneDimensionalElementaryWolframRuleModel, (
    index=rule_index, colors=colors, radius=radius
));
```

fill me in

```julia
base_frames = simulate(rulemodel, initial, steps=number_of_steps, algorithm  =  WolframDeterministicSimulation());
```

```julia
Gray.(1 .- base_frames[number_of_steps])
```

## Task 2: Probabilistic Cellular Automata
In this task, we will implement a probabilistic one-dimensional cellular automaton using a [modified version of Rule 30](https://en.wikipedia.org/wiki/Rule_30). We'll introduce randomness into the evolution process and observe how it affects the resulting patterns.


```julia
modified_frames = simulate(rulemodel, initial, steps=number_of_steps, maxnumberofmoves=(number_of_states - 80), 
    algorithm=WolframStochasticSimulation());
```

```julia
Gray.(1 .- modified_frames[number_of_steps])
```
