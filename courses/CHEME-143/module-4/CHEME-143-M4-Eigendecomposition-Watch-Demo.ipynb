# Eigendecomposition of Data and Systems
In this module, we will explore the concept of eigendecomposition, which is a powerful mathematical tool used in various fields, including data science, machine learning, and systems analysis. The key idea underlying eigendecomposition is to let the data speak for itself and to extract the most important features from the data. 

By the end of this module, you will be able to define and demonstrate mastery of the following key concepts:

* __Eigendecomposition__ allows us to decompose a matrix into its constituent parts, the [eigenvectors and eigenvalues](https://en.wikipedia.org/wiki/Eigenvalues_and_eigenvectors). These values help us understand the structure of the data or system represented by the matrix. We'll look at two approaches to estimate the [eigenvalues and eigenvectors](https://en.wikipedia.org/wiki/Eigenvalues_and_eigenvectors) of a matrix:
* __Power iteration method__ estimates the _largest_ eigenvalue/eigenvector pair. Given a _diagonalizable_ matrix $\mathbf{A}$ the power iteration algorithm will produce a number $\lambda$, which is the greatest (in absolute value) eigenvalue of $\mathbf{A}$ and a nonzero vector $\mathbf{v}$ which is a corresponding eigenvector of $\lambda$ such that $\mathbf{A}\mathbf{v} = \lambda\cdot\mathbf{v}$.
* __QR iteration__ is another approach to compute the eigendecomposition of the matrix $\mathbf{A}$. However, unlike power iteration, this approach will give all eigenvalues and eigenvectors of the matrix $\mathbf{A}$. The QR factorization algorithm relies on the [QR decomposition](https://en.wikipedia.org/wiki/QR_decomposition), which itself relies on the [Gram-Schmidt algorithm](https://en.wikipedia.org/wiki/Gram–Schmidt_process).

While we will explore these two approaches to compute the eigendecomposition of a matrix, most, if not all, computing platforms already have built-in functionality to do this computation. Most of the time, there is no need to reinvent the wheel (and your implementation will likely be worse in terms of space and time complexity).

This is going to be interesting, so let's get started!

## Eigendecomposition Fundamentals

For a square matrix $\mathbf{A}\in\mathbb{R}^{m\times{m}}$, the eigendecomposition finds scalar values $\lambda_j$ (eigenvalues) and corresponding vectors $\mathbf{v}_j$ (eigenvectors) such that:

$$\mathbf{A}\mathbf{v}_{j} = \lambda_{j}\mathbf{v}_{j}\qquad{j=1,2,\ldots,m}$$

The complete eigendecomposition can be written as:
$$\mathbf{A} = \mathbf{V}\mathbf{\Lambda}\mathbf{V}^{-1}$$

where:
- $\mathbf{V} = [\mathbf{v}_1, \mathbf{v}_2, \ldots, \mathbf{v}_m]$ is the matrix of eigenvectors (columns)
- $\mathbf{\Lambda} = \text{diag}(\lambda_1, \lambda_2, \ldots, \lambda_m)$ is the diagonal matrix of eigenvalues

### Special Properties of Symmetric Matrices

For symmetric real matrices $\mathbf{A} = \mathbf{A}^T$, eigendecomposition has particularly nice properties:

1. **Real eigenvalues**: All eigenvalues $\lambda_j$ are real numbers
2. **Orthogonal eigenvectors**: $\mathbf{v}_i^T\mathbf{v}_j = 0$ for $i \neq j$
3. **Orthonormal basis**: Normalized eigenvectors form an orthonormal basis: $\langle\hat{\mathbf{v}}_i,\hat{\mathbf{v}}_j\rangle = \delta_{ij}$

For symmetric matrices, the eigendecomposition becomes:
$$\mathbf{A} = \mathbf{Q}\mathbf{\Lambda}\mathbf{Q}^T$$
where $\mathbf{Q}$ is orthogonal ($\mathbf{Q}^T\mathbf{Q} = \mathbf{I}$).

### Geometric Interpretation

**Eigenvectors** represent the fundamental directions of the linear transformation defined by $\mathbf{A}$. They are the only directions that remain unchanged (up to scaling) when the transformation is applied.

**Eigenvalues** represent the scaling factors along their corresponding eigenvector directions:
- $\lambda > 1$: stretching along the eigenvector direction
- $0 < \lambda < 1$: compression along the eigenvector direction  
- $\lambda < 0$: reflection and scaling
- $\lambda = 0$: projection to a lower-dimensional space

### Applications and Importance

- **Data analysis**: Principal Component Analysis (PCA) uses eigendecomposition to find directions of maximum variance
- **Graph theory**: Spectral graph analysis uses eigenvalues to understand network structure
- **Optimization**: Eigenvalues determine whether a matrix is positive definite (all $\lambda_i > 0$)
- **Systems analysis**: Stability analysis of dynamical systems

The key insight is that eigendecomposition reveals the "natural coordinates" or principal directions of the data or system represented by the matrix.

### Method 1: Power iteration
The [power iteration method](https://en.wikipedia.org/wiki/Power_iteration) is an iterative algorithm to compute the largest eigenvalue and its corresponding eigenvector of a square (real) matrix; we'll consider only real-valued matrices here, but this approach can also be used for matrices with complex entries. 

> __Application:__ The most famous application of [power iteration](https://en.wikipedia.org/wiki/Power_iteration) is the [Google PageRank algorithm](https://epubs.siam.org/doi/10.1137/050623280)  which uses power iteration, utilizes the dominant eigenvalue and its corresponding eigenvector of a link connection matrix to assess the importance of web pages within a connection network.

How does the [power iteration method](https://en.wikipedia.org/wiki/Power_iteration) work?

__Phase 1: Eigenvector__: Suppose we have a real-valued square _diagonalizable_ matrix $\mathbf{A}\in\mathbb{R}^{m\times{m}}$ whose eigenvalues have the property $|\lambda_{1}|\geq|\lambda_{2}|\dots\geq|\lambda_{m}|$. Then, the eigenvector $\mathbf{v}_{1}\in\mathbb{C}^{m}$ which corresponds to the largest eigenvalue $\lambda_{1}\in\mathbb{C}$ can be (iteratively) estimated as:
$$
\mathbf{v}_{1}^{(k+1)} = \frac{\mathbf{A}\mathbf{v}_{1}^{(k)}}{\Vert \mathbf{A}\mathbf{v}_{1}^{(k)} \Vert}\quad{k=0,1,2\dots}
$$

where $\lVert \star \rVert$ denotes the [L2 (Euclidean) vector norm](https://mathworld.wolfram.com/L2-Norm.html). The [power iteration method](https://en.wikipedia.org/wiki/Power_iteration) converges to a value for the eigenvector as $k\rightarrow\infty$ when a few properties are true, namely, $|\lambda_{1}|/|\lambda_{2}| > 1$ (i.e., the dominant eigenvalue is well-separated), and we pick an appropriate initial guess for $\mathbf{v}_{1}$ (in our case, a random vector will work).

__Phase 2: Eigenvalue__: Once we have an estimate for the eigenvector $\hat{\mathbf{v}}_{1}$, we can estimate the corresponding eigenvalue $\hat{\lambda}_{1}$ using [the Rayleigh quotient](https://en.wikipedia.org/wiki/Rayleigh_quotient). This argument proceeds from the definition of the eigenvalues and eigenvectors. We know, from the definition of eigenvalue-eigenvector pairs, that:
$$
\mathbf{A}\hat{\mathbf{v}}_{1} - \hat{\lambda}_{1}\hat{\mathbf{v}}_{1}\simeq{0}
$$
where we use the $\simeq$ symbol because we don't have the true eigenvector $\mathbf{v}_{1}$, only an estimate of it. To solve this expression for the (estimated) eigenvalue $\hat{\lambda}_{1}$, we multiply through by the transpose of the eigenvector and solve for the eigenvalue:
$$
\hat{\lambda}_{1} \simeq \frac{\hat{\mathbf{v}}_{1}^{\top}\mathbf{A}\hat{\mathbf{v}}_{1}}{\hat{\mathbf{v}}_{1}^{\top}\hat{\mathbf{v}}_{1}} = \frac{\left<\mathbf{A}\hat{\mathbf{v}}_{1},\hat{\mathbf{v}}_{1}\right>}{\left<\hat{\mathbf{v}}_{1},\hat{\mathbf{v}}_{1}\right>}
$$
where $\left<\star,\star\right>$ denotes [an inner product](https://mathworld.wolfram.com/InnerProduct.html). 

Let's explore a pseudo-code implementation of the [power iteration method](https://en.wikipedia.org/wiki/Power_iteration) to estimate the largest eigenvalue and its corresponding eigenvector of a matrix $\mathbf{A}$.

__Initialization__: Given a real-valued square matrix $\mathbf{A}\in\mathbb{R}^{m\times{m}}$, we specify $\epsilon>0$ as the convergence tolerance, `maxiter` as the maximum number of iterations, an initial guess for the eigenvector $\mathbf{v}_{1}^{(0)}$ (we'll use a random vector), an iteration counter $k\gets{1}$, and a boolean variable $\texttt{converged}\gets{\texttt{false}}$ to indicate whether the algorithm has converged.

> __Parameter selection rules of thumb__:
> 
> * __Convergence tolerance__ $\epsilon$: A good starting point is $\epsilon = 10^{-6}$ to $10^{-8}$ for most practical applications. For higher precision requirements, use $\epsilon = 10^{-10}$ to $10^{-12}$. The choice depends on the desired accuracy and the condition number of the matrix.
> 
> * __Maximum iterations__ `maxiter`: This should scale with the matrix size and desired precision. A rule of thumb is:
>   - Small matrices ($m < 100$): `maxiter = 1000`
>   - Medium matrices ($100 \leq m < 1000$): `maxiter = 2000` 
>   - Large matrices ($m \geq 1000$): `maxiter = 5000`
>   
>   However, if the eigenvalue separation ratio $|\lambda_{1}|/|\lambda_{2}|$ is close to 1, you may need to increase these values significantly.
> 
> * __When to worry about convergence__: If the algorithm doesn't converge within the maximum iterations, this often indicates that the dominant eigenvalue is not well-separated from the second-largest eigenvalue, or the matrix may have complex eigenvalues with equal moduli.

Normalize the initial guess for the eigenvector: $\mathbf{v}_{1}^{(0)} \gets \frac{\mathbf{v}_{1}^{(0)}}{\lVert \mathbf{v}_{1}^{(0)} \rVert}$.

While not $\texttt{converged}$ __do__:
- Compute the matrix-vector product $\mathbf{y}^{(k)}\gets\mathbf{A}\mathbf{v}_{1}^{(k-1)}$.
- Normalize the vector $\mathbf{v}_{1}^{(k)}\gets{\mathbf{y}^{(k)}}/{\lVert \mathbf{y}^{(k)} \rVert}$.
- Compute the Rayleigh quotient to estimate the eigenvalue: $\hat{\lambda}_{1}^{(k)} \gets  {\mathbf{v}_{1}^{(k)\top}\mathbf{A}\mathbf{v}_{1}^{(k)}}/{\mathbf{v}_{1}^{(k)\top}\mathbf{v}_{1}^{(k)}}$.
- Check for convergence:
   1. If $\lVert \mathbf{v}_{1}^{(k)} - \mathbf{v}_{1}^{(k-1)} \rVert\leq\epsilon$, then set $\texttt{converged}\gets{\texttt{true}}$ and return $\hat{\lambda}_{1} = \hat{\lambda}_{1}^{(k)}$ and $\hat{\mathbf{v}}_{1} = {\mathbf{v}_{1}^{(k)}}/{\lVert \mathbf{v}_{1}^{(k)} \rVert}$.
   2. If $k\geq\texttt{maxiter}$, then set $\texttt{converged}\gets{\texttt{true}}$ to terminate the algorithm, return the last estimate of the eigenvalue and eigenvector (optionally flagging that convergence tolerance was not met).
   3. Increment $k\gets{k+1}$.


While simple to implement, the [power iteration method](https://en.wikipedia.org/wiki/Power_iteration) may exhibit slow convergence, mainly when the largest eigenvalue is close in magnitude to other eigenvalues, i.e., $|\lambda_{1}|/|\lambda_{2}| \sim 1$.

### Method 2: QR Iteration
[QR iteration](https://en.wikipedia.org/wiki/QR_algorithm) is a staple in numerical linear algebra for finding eigenvalues (and, with a small extension, eigenvectors).  Suppose we have a matrix $\mathbf{A}\in\mathbb{R}^{m\times{m}}$. The QR iteration algorithm is an iterative method that decomposes the matrix into orthogonal and upper triangular components. It repeatedly applies a QR decomposition:
$$
\mathbf{A}_k = \mathbf{Q}_k \mathbf{R}_k,
$$
where $\mathbf{Q}_k$ is an orthogonal matrix (i.e., $\mathbf{Q}_k^\top \mathbf{Q}_k = \mathbf{I}$) and $\mathbf{R}_k$ is an upper triangular matrix. The algorithm then forms a new matrix $\mathbf{A}_{k+1}$ by multiplying the $\mathbf{R}_k$ and $\mathbf{Q}_k$ matrices in reverse order:
$$
\mathbf{A}_{k+1} = \mathbf{R}_k\,\mathbf{Q}_k \;=\; \mathbf{Q}_k^\top \mathbf{A}_k\,\mathbf{Q}_k.
$$
Under mild conditions (e.g., real symmetric $\mathbf{A}$), $\mathbf{A}_k$ converges to an upper‐triangular matrix whose diagonal entries are the eigenvalues of the original $\mathbf{A}$.

**Algorithm**

1. **Initialization.**

   * Set $\mathbf{A}_0 = \mathbf{A}$.
   * Choose a maximum iteration count `maxiter` and tolerance $\epsilon$.

> __Parameter selection rules of thumb__:
> 
> * __Convergence tolerance__ $\epsilon$: For QR iteration, use tighter tolerances than power iteration since we're computing all eigenvalues. A good starting point is $\epsilon = 10^{-8}$ to $10^{-10}$ for most applications. For high precision requirements, use $\epsilon = 10^{-12}$ to $10^{-14}$.
> 
> * __Maximum iterations__ `maxiter`: QR iteration typically converges faster than power iteration but requires more computation per iteration. Rules of thumb:
>   - Small matrices ($m < 50$): `maxiter = 1000`
>   - Medium matrices ($50 \leq m < 200$): `maxiter = 2000`
>   - Large matrices ($m \geq 200$): `maxiter = 5000`
>   
>   For symmetric matrices, convergence is generally faster and these values can often be reduced by 20-30%.
> 
> * __When to worry about convergence__: QR iteration has better convergence guarantees than power iteration, especially for symmetric matrices. Slow convergence typically indicates ill-conditioned matrices or the presence of clustered eigenvalues.

2. **Iterate.**
   For $k = 0,1,2,\dots$ up to `maxiter-1`:

   * Compute the QR factorization $\mathbf{A}_k = \mathbf{Q}_k \mathbf{R}_k$.
   * Form the next iterate $\mathbf{A}_{k+1} \;\gets\; \mathbf{R}_k\,\mathbf{Q}_k
        \;=\; \mathbf{Q}_k^\top\,\mathbf{A}_k\,\mathbf{Q}_k$

3. **Convergence test.**
   Stop when either

   $$
     \|\mathbf{A}_{k+1} - \mathbf{A}_k\|_1 \le \epsilon
     \quad\text{or}\quad
     \|\lambda^{(k+1)} - \lambda^{(k)}\|_2 \le \epsilon,
   $$

   where $\lambda^{(k)}$ is the vector of diagonal entries of $\mathbf{A}_k$.


**Convergence guarantee:**  If $\mathbf{A}$ is a real symmetric (or complex Hermitian) matrix, then $\mathbf{A}_k\to \mathbf{T}$, an upper-triangular (in fact, diagonal) matrix whose diagonal entries are the real eigenvalues of $\mathbf{A}$. Once converged to $\mathbf{A}_\star$, its diagonal gives the eigenvalues.  To recover eigenvectors $\{\mathbf{v}_\star\}$, solve

$$
(\mathbf{A} - \lambda_\star \mathbf{I})\,\mathbf{v}_\star = 0
$$

for each eigenvalue $\lambda_\star$. Unlike power iteration (which finds only the dominant eigenpair), QR iteration delivers the full spectrum.

Wow, pretty cool, right? The QR iteration algorithm is a powerful method for computing eigenvalues and eigenvectors of matrices. However, before we implement it, how do we compute the $\mathbf{Q}$ and $\mathbf{R}$ matrices? Up next is a really handy algorithm called the [Gram-Schmidt Algorithm](https://en.wikipedia.org/wiki/Gram%E2%80%93Schmidt_process).

### Classical and Modified Gram–Schmidt
The QR decomposition of $\mathbf{A}\in\mathbb{R}^{m\times n}$ can be obtained by building an orthonormal basis $\{\mathbf{q}_1,\dots,\mathbf{q}_n\}$ for its columns via Gram–Schmidt orthogonalization.  Starting with the linearly independent columns $\mathbf{x}_1,\dots,\mathbf{x}_n$ of $\mathbf{A}$, we form the orthonormal basis vectors $\mathbf{q}_1,\dots,\mathbf{q}_n$ as:
$$
\mathbf{q}_k \;=\; \mathbf{x}_k \;-\;\sum_{i=1}^{k-1}c_{i,k}\,\mathbf{q}_i,
\quad
c_{i,k} = \mathbf{q}_i^\top\,\mathbf{x}_k,
$$
and then normalize $\mathbf{q}_k\gets \mathbf{q}_k/\|\mathbf{q}_k\|_2$.  Collecting the $\mathbf{q}$’s as columns of $\mathbf{Q}\in\mathbb{R}^{m\times n}$ and the $c_{i,k}$ into an upper-triangular $\mathbf{R}\in\mathbb{R}^{n\times n}$ gives $\mathbf{A}=\mathbf{Q}\,\mathbf{R}$.

#### Classical Gram-Schmidt
Let's look at how we can implement the classical Gram-Schmidt algorithm to compute the QR decomposition of a matrix $\mathbf{A}\in\mathbb{R}^{m\times{n}}$.

__Initialization__: Given a real-valued matrix $\mathbf{A}\in\mathbb{R}^{m\times{n}}$, initialize an empty matrix $\mathbf{Q} = \mathbf{0}\in\mathbb{R}^{m\times{n}}$ to hold the orthogonal vectors. 

For each column $j=1,2,\dots,n$ of the matrix $\mathbf{A}$ __do__:
- Set: $\mathbf{v}_{j} \gets \mathbf{A}_{j}$.
- For each previous column $k=1,2,\dots,j-1$ __do__:
    - Get the k-th column: $\mathbf{q}_{k} \gets \mathbf{Q}_{k}$.
    - Update: $\mathbf{v}_{j} \gets \mathbf{v}_{j} - c_{k,j}\cdot\mathbf{q}_{k}$, where $c_{k,j} = \mathbf{q}^{\top}_{k}\mathbf{v}_{j}$.
- Set (and normalize) the j-th column of the $\mathbf{Q}$ matrix: $\mathbf{Q}_{j} \gets \mathbf{v}_{j}/{\lVert \mathbf{v}_{j} \rVert_{2}}$.

Compute the $\mathbf{R}$ matrix: $\mathbf{R} \gets \mathbf{Q}^{\top}\;\mathbf{A}$.


__Cool, but there is a gotcha:__ Classical Gram-Schmidt can sometimes produce _almost_ orthogonal vectors because of roundoff error, which led to the Modified Gram-Schmidt algorithm. Let's see how it works.

#### Modified Gram-Schmidt
The [Modified Gram-Schmidt algorithm](https://en.wikipedia.org/wiki/Gram–Schmidt_process) is a variant of the classical Gram-Schmidt algorithm that improves numerical stability by orthogonalizing each vector against all previously computed vectors immediately, rather than waiting until the end of the process. This helps to reduce the accumulation of numerical errors that can occur in the classical version. Let's see how the modified Gram-Schmidt algorithm works.

__Initialization__: Given a real-valued matrix $\mathbf{A}\in\mathbb{R}^{m\times{n}}$, initialize an empty matrix $\mathbf{Q} = \mathbf{0}\in\mathbb{R}^{m\times{n}}$ to hold the orthogonal vectors. 

For each column $j=1,2,\dots,n$ of the matrix $\mathbf{A}$ __do__:
- Get the jth column: $\mathbf{v}_{j} \gets \mathbf{A}_{j}$.

For each column $j=1,2,\dots,n$ __do__:
- Set the j-th column of the $\mathbf{Q}$ matrix: $\mathbf{q}_{j} \gets \mathbf{v}_{j}/\lVert \mathbf{v}_{j} \rVert_{2}$.
- For $k = j+1,\dots,n$ __do__:
    - Update the k-th column: $\mathbf{v}_{k} \gets \mathbf{v}_{k} -\underbrace{\left( \mathbf{q}_{j}^{\top}\mathbf{v}_{k}\right)}_{c_{j,k}}\cdot\mathbf{q}_{j}$.

Compute the $\mathbf{R}$ matrix: $\mathbf{R} \gets \mathbf{Q}^{\top}\;\mathbf{A}$.


Both versions produce an orthonormal $\mathbf{Q}$ and triangular $\mathbf{R}$, but **Modified** Gram–Schmidt is generally preferred in practice because it better preserves $\mathbf{Q}^\top \mathbf{Q} = \mathbf{I}$ in finite precision.
