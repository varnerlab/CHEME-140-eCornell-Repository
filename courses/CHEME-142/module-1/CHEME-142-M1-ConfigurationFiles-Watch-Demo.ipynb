# Configuration Files
Hardcoding settings, like database passwords, file paths, or API keys, directly into your source code is a bad habit. It makes your code brittle, hard to share, and insecure.

This is where configuration files come in.

> __What are we looking at?__
>
> Instead of hardcoding, we use **Configuration Files** to store these settings separately. This allows you (or your users) to change how the program behaves without ever touching the actual code.
>
> In this demo, we'll look at the "Big Three" formats you'll encounter most often: **TOML**, **YAML**, and **JSON**. They are all human-readable, but each has its own quirks and best use cases.

Let's start with TOML.

___

## TOML (Tom's Obvious, Minimal Language)
First up is **TOML**. As the name suggests, it was designed to be obvious and minimal. It uses simple key-value pairs and brackets `[]` to define sections, making it look a lot like an old-school Windows `.INI` file.

Here is what a typical TOML config looks like:

```toml
# This is a TOML configuration file for a database

# section: holds connection information
[connection]
host = "localhost"      # The database hostname
port = 5432             # The port to connect to the database on
database = "mydatabase" # The name of the database
user = "myuser"         # The username to connect to the database with
password = "mypassword" # The password for the user
max_connections = 10    # The maximum number of connections to allow at once
connection_timeout = 30 # The amount of time to wait before timing out a connection

# section: holds a group of database options
[options]
ssl = true              # Whether to enable SSL connections to the database
ssl_mode = "require"    # The preferred SSL mode to use
```

You might be wondering why we are starting with TOML.

> __The Julia Connection__
>
> If you've been using Julia, you've already seen TOML! The `Project.toml` file that tracks your packages is, you guessed it, a TOML file.
>
> Because it is so central to the language, the `TOML` package is part of Julia's **Standard Library**. You don't need to install anything; just type `using TOML` and you're good to go.

Now, let's look at a format that is popular in the Python world: YAML.
___

## YAML (YAML Ain't Markup Language)
Next, we have **YAML**. It is very similar to TOML in that it's designed to be readable, but it uses **indentation** (whitespace) to define structure instead of brackets. If you come from a Python background, this will feel very familiar.

Here is that same configuration data, rewritten in YAML:

```yaml
# This is a YAML configuration file for an application

# Metadata about MyApp
name: MyApp             # The application's name
version: 1.0.0          # The version of the application
host: localhost         # The hostname to bind the application to
port: 8080              # The port to bind the application to

# A group of database options
database:
  host: localhost       # The database hostname
  port: 5432            # The port to connect to the database on
  name: mydatabase      # The name of the database
  user: myuser          # The username to connect to the database with
  password: mypassword  # The password for the user
```

However, this cleanliness comes with a catch.

> __A word of caution__
>
> YAML is clean and concise, but it is **extremely strict about indentation**. A single misplaced space or tab can break your entire file. Also, unlike TOML, YAML support is **not** built into Julia. You will need to add a third-party package like [`YAML.jl`](https://github.com/JuliaData/YAML.jl) to read these files.

Finally, let's look at the most common format for web data: JSON (JavaScript Object Notation).

___

## JSON (JavaScript Object Notation)
While TOML and YAML are great for humans writing configs, JSON is the standard language for **machines talking to machines**. It's a bit more verbose—lots of curly braces `{}` and quotes `""`—but it is universally understood by almost every programming language.

Here is a JSON example storing contact info:

```json
{
  "people": [
    {
      "name": "John Smith",
      "email": "john@example.com",
      "phone": "555-555-5555"
    },
    {
      "name": "Jane Smith",
      "email": "jane@example.com",
      "phone": "444-444-4444"
    }
  ]
}
```

Why is JSON so popular if it's harder to write?

> __The Universal Language__
>
> You will see JSON everywhere. It is the default format for web APIs, cloud services, and data exchange.
>
> *   **Structure:** It supports nested objects (dictionaries) and arrays (lists), making it very flexible.
> *   **Julia Support:** Like YAML, JSON is not in the standard library. However, the ecosystem has excellent packages like `JSON3.jl` that are blazing fast.

That covers the basics of configuration files. Now you can choose the right tool for the job!
