### Implementations
The notebook uses three helper functions to read different configuration file formats. Each function validates the file path and extension before parsing to ensure the correct format is being read.

> * `readtomlfile(path::String)`: Loads a TOML file from the specified path and returns a `Dict{String,Any}` containing the parsed configuration data. This function uses Julia's built-in `TOML.parsefile` method from the standard library.

> * `readjsonfile(path::String)`: Loads a JSON file from the specified path and returns a `Dict{String,Any}` containing the parsed data. This function uses the `JSON.parsefile` method from the `JSON.jl` package.

> * `readyamlfile(path::String)`: Loads a YAML file from the specified path and returns a `Dict{String,Any}` containing the parsed data. This function uses the `YAML.load_file` method from the `YAML.jl` package.

Here are the implementations of these three helper functions:

```julia

"""
    readjsonfile(path::String) -> Dict{String,Any}

Load the JSON file at the path arg. Returns a Dict{String,Any} 
containing the JSON data.
"""
function readjsonfile(path::String)::Dict{String,Any}

    # check: does path point to a json file?
    is_path_a_path = ispath(path);
    does_path_point_to_json_file = endswith(path, ".json");
    if (is_path_a_path == false) || (does_path_point_to_json_file == false)
        throw(ArgumentError("The path arg must point to a JSON file."))
    end

    # call the built-in parsefile function from the JSON package
    return JSON.parsefile(path)
end

"""
    readtomlfile(path::String) -> Dict{String,Any}

Load the TOML file at the path arg. Returns a Dict{String,Any} 
containing the TOML data.
"""
function readtomlfile(path::String)::Dict{String,Any}

    # check: does path point to a toml file?
    is_path_a_path = ispath(path);
    does_path_point_to_toml_file = endswith(path, ".toml");
    if (is_path_a_path == false) || (does_path_point_to_toml_file == false)
        throw(ArgumentError("The path arg must point to a TOML file."))
    end

    # call the built-in parsefile function from the TOML package
    return TOML.parsefile(path)
end

"""
    readyamlfile(path::String) -> Dict{String,Any}

Load the YAML file at the path arg. Returns a Dict{String,Any} 
containing the YAML data.
"""
function readyamlfile(path::String)::Dict{String,Any}

    # check: does path point to a yaml file?
    is_path_a_path = ispath(path);
    does_path_point_to_yaml_file = endswith(path, ".yaml");
    if (is_path_a_path == false) || (does_path_point_to_yaml_file == false)
        throw(ArgumentError("The path arg must point to a YAML file."))
    end

    return YAML.load_file(path)
end
```

## Example 1: Load TOML configuration files
[TOML (Tom's Obvious, Minimal Language)](https://toml.io/en/) is a configuration file format used to store application configuration data. It is intended to be easy to read, write, and parse. TOML files are widely used for storing configuration information. For example, in [Julia](https://docs.julialang.org), the [package manager Pkg.jl](https://pkgdocs.julialang.org/v1/) holds information about the packages required for a project in [a `Project.toml` file](Project.toml) (which is automatically created when a [project is activated](https://pkgdocs.julialang.org/v1/api/#Pkg.activate)).

> __TOML file structure__
>
> TOML files consist of key-value pairs, similar to a dictionary in Julia or Python, and can also include nested groups of keys. TOML files often have a `.toml` file extension.

Because of its central role in Julia, [the `TOML.jl` package](https://docs.julialang.org/en/v1/stdlib/TOML/) to read and write TOML files is included in the [Julia standard library](https://docs.julialang.org/en/v1/stdlib/TOML/). Thus, we don't need to install it and can access it by placing the `using TOML` statement at the start of our program.


### Example.toml
Let's load a sample TOML file and see how it works. We'll read the [Example.toml](data/Example.toml) file in the `data` directory. First, let's specify a `path` to the file we want to load, i.e., a `String` that tells our program where to find the file we want to read.

> __Why use `joinpath`?__
>
> In Julia, we use the [joinpath function](https://docs.julialang.org/en/v1/base/file/#Base.Filesystem.joinpath) to specify paths because this ensures that we use the proper directory separator character (`/` on macOS/Linux and `\` on Windows).

Let's build the path to our TOML file:

```julia
path_to_example_toml_file = joinpath(_PATH_TO_DATA, "Example.toml")
```

Now let's parse the TOML file by calling the `readtomlfile(...)` function with the `path_to_example_toml_file::String` path. The function returns a `Dict{String,Any}` dictionary containing the parsed configuration data:

```julia
toml_dictionary = readtomlfile(path_to_example_toml_file) # call our helper function to parse the file
```

What's in the `toml_dictionary::Dict{String, Any}` dictionary? Let's check:

```julia
toml_dictionary["options"] # let's look at the options section
```

Now that we've loaded the TOML file, let's modify a value and write it back to disk. This demonstrates how to update configuration files programmatically.

> __The `open ... do` pattern__
>
> The [open function do pattern](https://docs.julialang.org/en/v1/base/io-network/#Base.open) provides a safe way to work with files. It automatically opens the file, executes the code block, and then closes the file—even if an error occurs. We use [TOML.print](https://docs.julialang.org/en/v1/stdlib/TOML/#TOML.print) to write the dictionary back to the file.

Let's check a value before we change it:

```julia
toml_dictionary["options"]["ssl_mode"]
```

```julia
TOML.print(toml_dictionary); # 
```

```julia
toml_dictionary["options"]["ssl"] = true;
open(path_to_example_toml_file,"w") do io
    TOML.print(io,toml_dictionary)
end
```

The [open function do pattern in Julia](https://docs.julialang.org/en/v1/base/io-network/#Base.open) is an evolution of [traditional file handling in C](https://www.freecodecamp.org/news/file-handling-in-c-how-to-open-close-and-write-to-files/). It is generic and can be used to write (or read, as we shall see) many different types of files.

Here's an example of using it to read a file line by line:

```julia
let
    path_to_data_file = path_to_example_toml_file;
    open(path_to_data_file, "r") do io # open a stream to the file
        for line ∈ eachline(io) # read each line from the file
            println(line) # print the line to the console
        end
    end
end
```

## Example 2: JSON configuration files
[JavaScript Object Notation (JSON)](https://developer.mozilla.org/en-US/docs/Learn/JavaScript/Objects/JSON) is a lightweight, text-based, language-independent data interchange format that is easy for humans to read and write and easy for machines to parse and generate. JSON is based on a subset of the [JavaScript programming language](https://en.wikipedia.org/wiki/JavaScript) and represents simple data structures and associative arrays.

> __JSON data structures__
>
> JSON is composed of two data structures: a collection of name/value pairs typically realized as a struct, dictionary, keyed list, or associative array; and an ordered list of values, typically realized as an array, vector, list, or sequence.

Unlike TOML, the JSON format is not included in the [Julia standard library](https://docs.julialang.org/en/v1/stdlib/TOML/). Instead, a variety of third-party packages are available for reading and writing JSON files, e.g., the [JSON.jl](https://github.com/JuliaIO/JSON.jl) package.

### Example.json
Let's load a sample JSON file to see how it works. We'll read the [Example.json](data/Example.json) file in the `data` directory. As before, we'll use `joinpath` to construct the path.

> __Why use `joinpath`?__
>
> In Julia, we use the [joinpath function](https://docs.julialang.org/en/v1/base/file/#Base.Filesystem.joinpath) to specify paths because this ensures that we use the proper directory separator character (`/` on macOS/Linux and `\` on Windows).

Let's build the path and load the JSON file:

```julia
path_to_example_json_file = joinpath(_PATH_TO_DATA, "Example.json")
```

Now let's parse the JSON file by calling the `readjsonfile(...)` function with the `path_to_example_json_file::String` path. The function returns a `Dict{String,Any}` dictionary containing the parsed JSON data:

## Example 3: YAML configuration files
[YAML (YAML Ain't Markup Language)](https://yaml.org) is a human-readable data serialization language used to transmit data between systems. Like TOML, YAML is often used as an application configuration file format.

> __YAML file structure__
>
> YAML files use a simple syntax that consists of key-value pairs and can also include nested groups of keys. YAML uses indentation to denote structure, similar to [Python](https://www.python.org). YAML files often have a `.yaml` or `.yml` file extension.

Unlike TOML, the YAML format is not included in the [Julia standard library](https://docs.julialang.org/en/v1/stdlib/TOML/). Instead, various third-party packages are available for working with YAML files, e.g., the [YAML.jl](https://github.com/JuliaData/YAML.jl) package.

### Example.yaml
Let's load a sample YAML file to complete our tour of configuration formats. We'll read the [Example.yaml](data/Example.yaml) file in the `data` directory. As with the previous examples, we'll use `joinpath` to construct the path.

> __Why use `joinpath`?__
>
> In Julia, we use the [joinpath function](https://docs.julialang.org/en/v1/base/file/#Base.Filesystem.joinpath) to specify paths because this ensures that we use the proper directory separator character (`/` on macOS/Linux and `\` on Windows).

Let's build the path and load the YAML file:

```julia
path_to_example_yaml_file = joinpath(_PATH_TO_DATA, "Example.yaml")
```

Now let's parse the YAML file by calling the `readyamlfile(...)` function with the `path_to_example_yaml_file::String` path. The function returns a `Dict{String,Any}` dictionary containing the parsed YAML data:

```julia
yaml_dictionary = readyamlfile(path_to_example_yaml_file)
```

What's in the `yaml_dictionary::Dict{String, Any}`? Let's check:

```julia
yaml_dictionary["database"]
```

## Summary
This example demonstrated how to read and parse configuration files in three common formats: TOML, JSON, and YAML.

> __Key Takeaways:__
> 
> * **Format-specific parsers:** Each configuration file format requires a specific parser—TOML uses Julia's standard library, while JSON and YAML require third-party packages to convert file contents into Julia dictionaries.

Configuration files enable external storage of application settings, making programs more flexible and easier to maintain without modifying source code.
