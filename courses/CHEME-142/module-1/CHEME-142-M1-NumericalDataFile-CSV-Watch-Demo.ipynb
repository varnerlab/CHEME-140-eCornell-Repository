# Numerical Data Files
Numerical data files are used to store numerical information in a structured way. These files can hold data from experiments, simulations, or other sources. Numerical data files might come in different formats, such as comma-separated values (CSV), space-separated values (SSV), tab-separated values (TSV), or other delimited formats.

Let's examine a typical comma-separated values (CSV) file that stores data; similar ideas can be applied to other delimited formats like SSV or TSV files.

## CSV files
[Comma-separated values (CSV)](https://en.wikipedia.org/wiki/Comma-separated_values) files are _delimited_ text files that use commas to separate values; a space or tab character can also serve as a delimiter. CSV files are widely used for storing and exchanging numerical data, especially in engineering and scientific fields. 

> __What is the structure of a CSV file?__ A CSV file typically contains tabular data (numbers and text) in plain text, with each line having the same number of fields (although this is not always guaranteed). Each line in the file is called a _record_, and each record contains one or more _fields_, separated by commas (or other characters such as a `tab` or `space`).

In engineering or other quantitative applications, [comma-separated values files](https://en.wikipedia.org/wiki/Comma-separated_values) are typically used to store, transmit, and work with numerical data. Consider a comma-separated values file holding historical interest rate data:

```CSV
Date,T=20-year-percentage,T=30-year-percentage
2021-09-17,1.82,1.88
2021-09-24,1.84,1.89
2021-10-01,2.00,2.05
2021-10-08,2.05,2.10
....
```

> __What are we looking at?__
>
> First, notice the **header** in the top row. It tells us what each column means: dates and interest rates. While headers aren't strictly required, they make life much easier! Without them, you'd need a separate "data dictionary" just to figure out what the numbers represent.
>
> Below the header, we have the **data records**. Each row is a single record, and the values (fields) are separated by commas. It's organized and predictable, every row usually has the same number of items.

However, while the data itself is clear, CSV files don't tell the whole story.

> __What is missing?__
>
> You might notice there's no **metadata**. The file doesn't tell us the units (is it percent? dollars?) or where the data came from. That context usually lives in a separate document, though sometimes you'll see comments (lines starting with `#`) tucked inside the file.
>
> Also, CSVs are **flat**. They can't easily handle complex, nested structures like a JSON file can. If one piece of data depends on another in a complicated way, a simple list of rows and columns might not cut it.

Beyond these structural limitations, there are also some practical details to keep in mind when working with real-world files.

> __A few things to watch out for__
>
> *   **Commas in the data:** What if a field contains a comma, like "New York, NY"? To prevent confusion, we wrap that field in quotes: `"New York, NY"`.
> *   **Missing values:** Real data is rarely perfect. Missing items might show up as two commas side-by-side `,,`, or special markers like `NaN` or `NULL`.
> *   **Encoding:** Since CSVs are just text, the computer needs to know how to read the characters. `UTF-8` is the standard encoding that handles most languages and symbols without turning them into gibberish.

Finally, do we always have to use commas as delimiters in CSV files? No! Other characters, such as spaces or tabs, can also be used as delimiters. When a different delimiter is used, the file is often referred to as a "delimited text file" rather than a CSV file. For example, a tab-separated values (TSV) file uses tabs as delimiters instead of commas.
