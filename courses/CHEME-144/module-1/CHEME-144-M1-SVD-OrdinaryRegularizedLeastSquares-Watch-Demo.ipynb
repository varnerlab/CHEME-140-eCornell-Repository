\end{equation*}

\begin{equation*}
\hat{\mathbf{\beta}} = \mathbf{V}\cdot\mathbf{\Sigma}^{-1}\cdot\mathbf{U}^{T}\cdot\mathbf{y}
\end{equation*}
or in index notation as:
\begin{equation*}
\hat{\beta} = \sum_{i=1}^{r}\frac{\mathbf{u}_{i}^{T}\cdot\mathbf{y}}{\sigma_{i}}\cdot\mathbf{v}_{i}
\end{equation*}
where $r$ denotes the rank of the data matrix $\mathbf{X}$, $\mathbf{u}_{i}$ and $\mathbf{v}_{i}$ are the $i$-th columns of $\mathbf{U}$ and $\mathbf{V}$, respectively,
