# Introduction to Linear Least Squares and Regression
In this module, we will explore the concept of linear least squares and regression, which are fundamental techniques in machine learning and statistics. The goal is to understand how to model relationships between input features and target variables using linear models.

By the end of this module, you will be able to define and demonstrate mastery of the following key concepts:

* __Linear regression__ is a statistical method for modeling the relationship between a target variable (output) and one or more features (input) by fitting a linear model to observed data. It provides a straightforward method for predicting outcomes and understanding the relationships between variables. The model is linear in the parameters, not necessarily in the features.
* __Continuous variable prediction tasks__: In machine learning, linear regression models are commonly employed for continuous variable prediction tasks. These models enable the estimation of numerical outcomes based on the (non)linear relationships identified between input features and the target variable.
 

You’ll master linear regression for modeling feature–outcome relationships and predicting continuous variables across a variety of tasks. You’ll also learn how to extend these models to binary classification by transforming numerical predictions into discrete class estimates. Let's get started!

## Linear models for continuous prediction tasks
Suppose there exists a dataset $\mathcal{D} = \left\{(\mathbf{x}_{i},y_{i}) \mid i = 1,2,\dots,n\right\}$ with $n$-training (labeled) examples, where $\mathbf{x}_{i}\in\mathbb{R}^{m}$ is a $m$-dimensional vector of features (independent input variables, typically real but potentially complex) and $y_{i}\in\mathbb{R}$ denotes a scalar response variable (dependent variable). Then, a $\texttt{linear model}$ for the dataset $\mathcal{D}$ is given (in index-form) by:
$$
\begin{equation*}
y_{i} = \hat{\mathbf{x}}_{i}^{\top}\,\mathbf{\beta} + \epsilon_{i}\qquad{i=1,2,\dots,n}
\end{equation*}
$$
where the features are $\hat{\mathbf{x}}^{\top}=\left(x_{1},x_{2},\dots,x_{m},1\right)$ (we've added an extra `1` to each feature vector to account for the intercept (bias) term), 
the unknown parameters are represented by the $\mathbf{\beta}\in\mathbb{R}^{p}$ vector (where $p=m+1$), and $\epsilon_{i}\in\mathbb{R}$ is the unobserved random error for response $i$, i.e., the component of the target that is _not_ explained by the linear model. We can rewrite the linear regression model in matrix-vector form as:
$$
\begin{equation*}
\mathbf{y} = \hat{\mathbf{X}}\;\mathbf{\beta} + \mathbf{\epsilon}
\end{equation*}
$$
where $\hat{\mathbf{X}}$ is an $n\times{p}$ matrix with the features $\hat{\mathbf{x}}_{i}^{\top}$ on the rows, the target (output) vector $\mathbf{y}$ is an $n\times{1}$ column vector with entries $y_{i}$, and the error vector $\mathbf{\epsilon}$ is an $n\times{1}$ column vector with entries $\epsilon_{i}$. The challenge of linear regression is to estimate the unknown parameters $\mathbf{\beta}$ from the dataset $\mathcal{D}$ such that we minimize some objective function.

__Interesting__: A linear model must only be linear in the parameters, not necessarily the features. For example, we could have rows in the data matrix $\mathbf{X}$ that were non-linear in the features, e.g., $1,x,x^{2},x^{3},\dots$. This would still be a linear regression problem because the model is linear in the parameters.

___

## Overdetermined data matrix without regularization
Suppose you have a data matrix $\hat{\mathbf{X}}\in\mathbb{R}^{n\times{p}}$ that is $\texttt{overdetermined}$, i.e., $n \gg p$, and an error model $\mathbf{\epsilon}\sim\mathcal{N}(\mathbf{0},\sigma^{2}\;\mathbf{I})$ follows [a Normal distribution](https://en.wikipedia.org/wiki/Normal_distribution) with a mean of zero and variance $\sigma^{2}$. We estimate the model parameters by minimizing the sum of squared errors between the model's estimated outputs and the observed outputs:
$$
\begin{align*}
\hat{\mathbf{\beta}} = \arg\min_{\mathbf{\beta}} \frac{1}{2}\;\lVert~\mathbf{y} - \hat{\mathbf{X}}\;\mathbf{\beta}~\rVert^{2}_{2}
\end{align*}
$$
where $\lVert\star\rVert^{2}_{2}$ is the square of the [L2 vector norm](https://en.wikipedia.org/wiki/Norm_(mathematics)#Euclidean_norm), and $\hat{\mathbf{\beta}}\in\mathbb{R}^{p}$ is the estimated parameter vector. This problem has an analytical solution (take the gradient of the objective function with respect to the parameters, set to zero):
$$
\begin{align*}
\hat{\mathbf{\beta}} &= \left(\hat{\mathbf{X}}^{\top}\hat{\mathbf{X}}\right)^{-1}\hat{\mathbf{X}}^{\top}\mathbf{y}
\end{align*}
$$
However, we can also express the parameters in terms of the error model $\mathbf{\epsilon}$:
$$
\begin{align*}
\hat{\mathbf{\beta}} &= \left(\hat{\mathbf{X}}^{\top}\hat{\mathbf{X}}\right)^{-1}\hat{\mathbf{X}}^{\top}\mathbf{y} \\
\hat{\mathbf{\beta}}&= \left(\hat{\mathbf{X}}^{\top}\hat{\mathbf{X}}\right)^{-1}\hat{\mathbf{X}}^{\top}(\hat{\mathbf{X}}\;\mathbf{\beta} + \mathbf{\epsilon}) \\
\hat{\mathbf{\beta}} &= \underbrace{\left[\left(\hat{\mathbf{X}}^{\top}\hat{\mathbf{X}}\right)^{-1}\hat{\mathbf{X}}^{\top}\hat{\mathbf{X}}\right]}_{= \mathbf{I}}\;\mathbf{\beta} + \left(\hat{\mathbf{X}}^{\top}\hat{\mathbf{X}}\right)^{-1}\hat{\mathbf{X}}^{\top}\mathbf{\epsilon}\\
\hat{\mathbf{\beta}} &= \mathbf{\beta} + \left(\hat{\mathbf{X}}^{\top}\hat{\mathbf{X}}\right)^{-1}\hat{\mathbf{X}}^{\top}\mathbf{\epsilon}\quad\blacksquare\\
\end{align*}
$$

__Hmmm__: What is this saying to me?

* _Is it unique?_ Is there a unique solution? No, there is not! There is not a single parameter vector $\beta$ that satisfies all the equations simultaneously. We found the minimum norm solution; however, if we chose a different norm, we may have a different answer. 
* _Random variable?_ The $\hat{\mathbf{\beta}}$ expression is a random variable that looks like a _baby step_ towards [Bayesian linear regression](https://en.wikipedia.org/wiki/Bayesian_linear_regression). Why? Because $\mathbf{\epsilon}$ is a probability distribution function! Thus, we have a _distribution_ of possible parameters, not a single set.
* _Frequentist perspective?_ Traditionally, STEM folks come from the frequentist camp, assuming a single physical value for constants like a rate constant. But this analysis shows that including model error challenges this. The main difference is that Bayesians treat regression coefficients as random variables with distributions, unlike frequentists, who see them as fixed but unknown.

Let's look at the case where we have _limited data_ compared to the number of parameters in the model, i.e., $n\ll{p}$. This case is called $\texttt{underdetermined}$.
___

## Underdetermined data matrix without regularization
Next, let's assume the data matrix $\mathbf{X}$ is $\texttt{underdetermined}$, i.e., $n \ll p$, and the error model $\mathbf{\epsilon}\sim\mathcal{N}(\mathbf{0},\sigma^{2}\mathbf{I})$.
The ordinary least squares estimate of the unknown parameters $\mathbf{\beta}$ is the _smallest_ $\beta$ that satisfies the original equations:
$$
\begin{align*}
\text{minimize}~&  ||\,\mathbf{\beta}\,||^{2}_{2} \\
\text{subject to} & \, \hat{\mathbf{X}}\;\mathbf{\beta} = \mathbf{y}
\end{align*}
$$
The least-norm solution of the unknown parameter vector is given by:
$$
\begin{align*}
\hat{\mathbf{\beta}} &= \hat{\mathbf{X}}^{\top}\left(\hat{\mathbf{X}}\hat{\mathbf{X}}^{\top}\right)^{-1}\;\mathbf{y}
\end{align*}
$$
We can express the solution in terms of the error model $\mathbf{\epsilon}$:
$$
\begin{align*}
\hat{\mathbf{\beta}} &= \hat{\mathbf{X}}^{\top}\left(\hat{\mathbf{X}}\hat{\mathbf{X}}^{\top}\right)^{-1}\;\mathbf{y}\\
\hat{\mathbf{\beta}} &= \hat{\mathbf{X}}^{\top}\left(\hat{\mathbf{X}}\hat{\mathbf{X}}^{\top}\right)^{-1}\;(\hat{\mathbf{X}}\;\mathbf{\beta} + \mathbf{\epsilon})\\
\hat{\mathbf{\beta}} &= \underbrace{\hat{\mathbf{X}}^{\top}\left(\hat{\mathbf{X}}\hat{\mathbf{X}}^{\top}\right)^{-1}\;\hat{\mathbf{X}}}_{\text{Projection}\;\mathbf{P}}\;\mathbf{\beta} + \hat{\mathbf{X}}^{\top}\left(\hat{\mathbf{X}}\hat{\mathbf{X}}^{\top}\right)^{-1}\;\mathbf{\epsilon}\\
\hat{\mathbf{\beta}} &= \mathbf{P}\;\mathbf{\beta} +\hat{\mathbf{X}}^{\top}\left(\hat{\mathbf{X}}\hat{\mathbf{X}}^{\top}\right)^{-1}\;\mathbf{\epsilon}\quad\blacksquare\\
\end{align*}
$$

__Hmmm__: A few questions.

* _Why do we select the smallest $\hat{\mathbf{\beta}}$?_ There is no unique solution, i.e., there are infinitely many solutions to the underdetermined system of equations. The least-norm solution is the one that minimizes the L2 norm of the parameter vector $\hat{\mathbf{\beta}}$, which is connected to the concept of regularization. 
* _What is Regularization?_ Regularization is a method used to prevent overfitting in machine learning models by adding a penalty to the loss function. In linear regression, regularization helps manage the model's complexity and enhances its ability to generalize to new data. Common techniques include Lasso (L1 regularization) and Ridge (L2 regularization). 

Let's revisit the overdetermined case and see how regularization can be applied to improve the model's performance.
___

## Regularized linear regression
In the overdetermined case, we can add a regularization term to the objective function to prevent overfitting and improve generalization. The regularized linear regression problem is given by:
$$
\begin{align*}
\hat{\mathbf{\beta}} = \arg\min_{\mathbf{\beta}}\left( \frac{1}{2}\;\lVert~\mathbf{y} - \hat{\mathbf{X}}\;\mathbf{\beta}~\rVert^{2}_{2} + \frac{\lambda}{2}\;\lVert~\mathbf{\beta}~\rVert^{2}_{2}\right)
\end{align*}
$$
where $\lambda> 0$ is the regularization parameter controlling regularization strength. This is Ridge (or Tikhonov) regularization. The first term sums squared errors, the second penalizes large parameters. Its solution is given by:
$$
\begin{align*}
\hat{\mathbf{\beta}} &= \left(\hat{\mathbf{X}}^{\top}\hat{\mathbf{X}} + \lambda\;\mathbf{I}\right)^{-1}\hat{\mathbf{X}}^{\top}\mathbf{y}
\end{align*}
$$
This solution can also be expressed in terms of the error model $\mathbf{\epsilon}$:
$$
\begin{align*}
\hat{\mathbf{\beta}} &= \left(\hat{\mathbf{X}}^{\top}\hat{\mathbf{X}} + \lambda\;\mathbf{I}\right)^{-1}\hat{\mathbf{X}}^{\top}(\hat{\mathbf{X}}\;\mathbf{\beta} + \mathbf{\epsilon}) \\
\hat{\mathbf{\beta}} &= \left(\hat{\mathbf{X}}^{\top}\hat{\mathbf{X}} + \lambda\;\mathbf{I}\right)^{-1}\hat{\mathbf{X}}^{\top}\hat{\mathbf{X}}\;\mathbf{\beta} + \left(\hat{\mathbf{X}}^{\top}\hat{\mathbf{X}} + \lambda\;\mathbf{I}\right)^{-1}\hat{\mathbf{X}}^{\top}\mathbf{\epsilon} \\
\hat{\mathbf{\beta}} &= \underbrace{\left(\hat{\mathbf{X}}^{\top}\hat{\mathbf{X}} + \lambda\;\mathbf{I}\right)^{-1}\hat{\mathbf{X}}^{\top}\hat{\mathbf{X}}}_{\text{Shrinkage}\;\mathbf{P}}\;\mathbf{\beta} + \left(\hat{\mathbf{X}}^{\top}\hat{\mathbf{X}} + \lambda\;\mathbf{I}\right)^{-1}\hat{\mathbf{X}}^{\top}\mathbf{\epsilon} \\
\hat{\mathbf{\beta}} &= \mathbf{P}\;\mathbf{\beta} + \left(\hat{\mathbf{X}}^{\top}\hat{\mathbf{X}} + \lambda\;\mathbf{I}\right)^{-1}\hat{\mathbf{X}}^{\top}\mathbf{\epsilon}\quad\blacksquare\\
\end{align*}
$$

__Interesting__: The regularization term $\lambda\;\mathbf{I}$ acts as a penalty for large parameter values, effectively shrinking the estimated parameters towards zero. This helps to prevent overfitting by discouraging complex models that fit the training data too closely.

Now, the question is: how do we select the regularization parameter $\lambda$? The answer is that we can use a validation set to tune the hyperparameter $\lambda$ by evaluating the model's performance on completely unseen data.
___

## A note on the error model
In the context of linear regression, the error model $\mathbf{\epsilon}$ is typically assumed to follow a Normal distribution with mean zero and variance $\sigma^{2}$:
$$\begin{align*}
\mathbf{\epsilon} &\sim \mathcal{N}(\mathbf{0},\sigma^{2}\;\mathbf{I}) \\
\end{align*}$$
This Normality assumption not only yields the closed‐form OLS/ridge solution but also underpins statistical inference (confidence intervals, hypothesis tests, etc.). (Useful! But with the caveat: the Normality assumption may not always hold!)

We can estimate the variance of the error model from the residuals $\mathbf{r} = \mathbf{y} - \hat{\mathbf{X}}\hat\beta$, which are the differences between the observed outputs and the predicted outputs:
$$\begin{align*}
\hat{\sigma}^{2} &= \frac{1}{n-p}\;\lVert~\underbrace{\mathbf{y} - \hat{\mathbf{X}}\;\hat{\mathbf{\beta}}}_{\text{residual}\;\mathbf{r}}~\rVert^{2}_{2}\\
&= \frac{1}{n-p}\;\lVert~\mathbf{r}~\rVert^{2}_{2}\\
\end{align*}$$
where $n$ is the number of training examples, $p$ is the number of parameters (including the intercept), and $\hat{\mathbf{\beta}}$ is the estimated parameter vector.
* _Unbiased estimator_: This estimate is known as the unbiased estimator of the noise variance (sometimes called the residual variance). However, if you divide by $n$ instead of $n-p$, you obtain the in‐sample mean squared error.

Now that we have a model for the error, we can use it to make predictions and evaluate the model's performance, as well as the uncertainty of the model parameters. 
Under the Gauss–Markov conditions and (approximate) normality, $\hat\beta$ is the **best linear unbiased estimator (BLUE)** of the true coefficients. You can then compute standard errors via:
$$
  \mathrm{SE}(\hat\beta_j) 
  = \sqrt{\;\hat\sigma^2\;\bigl[(\hat{\mathbf{X}}^\top\hat{\mathbf{X}})^{-1}\bigr]_{jj}\,}
$$
which are useful for computing prediction intervals, hypothesis tests, and other inferential tools. Cool!
