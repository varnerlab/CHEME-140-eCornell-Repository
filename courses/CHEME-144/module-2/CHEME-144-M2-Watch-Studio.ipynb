## What is Linear Programming?
Linear programming (LP) is an optimization framework for finding the best value of a linear objective function subject to linear constraints and non-negativity constraints.

Linear programming (LP) is a mathematical method for determining a way to achieve the best outcome in a given mathematical model whose requirements are represented by linear relationships. Let's look at the key components of the _primal_ linear programming problem.

Let $O:\mathbb{R}^{m}\to\mathbb{R}$ denote a _linear function_ of the continuous non-negative decision variables $\mathbf{x}\in\mathbb{R}^{m}$
whose values are constrained by a system of linear equations and bounded. Then, the optimal $\mathbf{x}^{\star}$ is a solution of the linear program:
$$
\begin{align*}
\text{maximize} &\, \sum_{i=1}^{m} c_{i}\cdot{x}_{i}\\
\text{subject to}~\mathbf{A}\cdot\mathbf{x} &\leq \mathbf{b}\\
\text{and}~x_{i}&\geq {0}\qquad{i=1,2,\dots,m}
\end{align*}
$$
The constants $c_{i}$ are coefficients in the objective function, $\mathbf{A}\in\mathbb{R}^{n\times{m}}$ 
is the constraint matrix and $\mathbf{b}\in\mathbb{R}^{n}$ is a constant vector. Any linear program (using some algebraic magic) can be converted into this standard form.

## How do we solve linear programs?
The most common method for solving linear programs is the Simplex algorithm, which iteratively moves along the edges of the feasible region defined by the constraints to find the optimal solution at a vertex (corner point). The Simplex algorithm is efficient for most practical problems, although it can take exponential time in the worst case. Let's try the most obvious approach first: the Lagrange multiplier method.

### Lagrange Multipliers
Normally, when faced with a constrained optimization problem, we'd use the Lagrange multiplier method to convert it into an unconstrained problem, and then find the optimal solution by computing the gradient of the Lagrangian function and setting it to zero. However, this method is not applicable to linear programming problems because the feasible region is defined by linear constraints, which do not allow for the use of Lagrange multipliers in the same way as nonlinear constraints. Let's check the Lagrangian function for a linear program.

Suppose we have a linear program in the form:
$$
\begin{align*}
\text{maximize} &\, \sum_{i=1}^{m} c_{i}\cdot{x}_{i}\\
\text{subject to}~\mathbf{A}\cdot\mathbf{x} &\leq \mathbf{b}\\
\text{and}~x_{i}&\geq {0}\qquad{i=1,2,\dots,m}
\end{align*}
$$
The first thing we do is to convert the inequality constraints into equality constraints by introducing slack variables $s_{j}$ for each constraint:
$$\begin{align*}
\text{maximize} &\, \sum_{i=1}^{m} c_{i}\cdot{x}_{i}\\
\text{subject to}~ \mathbf{A}\cdot\mathbf{x} + \mathbf{s} &= \mathbf{b}\\
x_{i} &\geq 0\quad{i=1,2,\dots,m}\\
s_{j} &\geq 0\quad{j=1,2,\dots,n}
\end{align*}$$
Now, we can write the Lagrangian function $\mathcal{L}(\mathbf{x}, \mathbf{s}, \boldsymbol{\lambda})$ as:
$$
\begin{align*}
\mathcal{L}(\mathbf{x}, \mathbf{s}, \boldsymbol{\lambda}) &= \mathbf{c}^{\top}\mathbf{x} - \boldsymbol{\lambda}^{\top}(\mathbf{A}\cdot\mathbf{x} + \mathbf{s} - \mathbf{b})\\
\end{align*}
$$
To compute the first-order conditions, we take the gradient of the Lagrangian with respect to $\mathbf{x}$, $\mathbf{s}$ and $\boldsymbol{\lambda}$ and set it to zero:
$$
\begin{align*}
\nabla_{\mathbf{x}}\mathcal{L}(\mathbf{x}, \mathbf{s}, \boldsymbol{\lambda}) &= \mathbf{c} - \mathbf{A}^{\top}\boldsymbol{\lambda} = 0\\
\nabla_{\mathbf{s}}\mathcal{L}(\mathbf{x}, \mathbf{s}, \boldsymbol{\lambda}) &= -\boldsymbol{\lambda} = 0\\
\nabla_{\boldsymbol{\lambda}}\mathcal{L}(\mathbf{x}, \mathbf{s}, \boldsymbol{\lambda}) &= \mathbf{A}\cdot\mathbf{x} + \mathbf{s} - \mathbf{b} = 0
\end{align*}
$$
From the second equation, we see that $\boldsymbol{\lambda} = 0$, which means that the Lagrange multipliers are zero for all constraints. This implies that the Lagrangian function does not provide any useful information about the optimal solution of the linear program. Further, this requires $\mathbf{c}= 0$, which is not generally true. Therefore, the Lagrange multiplier method is not applicable to linear programming problems. Hmmmm. 

### Karush–Kuhn–Tucker (KKT) Conditions
The Karush–Kuhn–Tucker (KKT) conditions extend the Lagrange multiplier framework to problems with inequality and non-negativity constraints, providing a unified set of stationarity, feasibility, and complementary-slackness requirements. In convex settings—LPs in particular—these conditions are both necessary and sufficient for optimality.
