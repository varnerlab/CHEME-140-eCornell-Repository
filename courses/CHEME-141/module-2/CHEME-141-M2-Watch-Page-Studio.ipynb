# Program Control Flow and Iteration
In this module, we explore ideas in control flow and iteration, focusing on how to structure programs to make decisions and repeat actions.

By the end of this module, you will be able to define and demonstrate mastery of the following key concepts:

* __If-Else Statements:__ Conditional statements that execute different code blocks based on whether a condition is true or false, allowing for branching logic in programs.
* __For Loops:__ Iterative constructs that repeat a block of code a specified number of times, often used for traversing collections or performing repeated calculations when the number of iterations is known in advance.
* __While Loops:__ Iterative constructs that repeat a block of code as long as a specified condition remains true, allowing for dynamic iteration where the number of iterations is not predetermined.

If-else statements and iteration patterns are foundational to programming. These structures exist in virtually all programming languages and provide the building blocks for more complex logic and functionality. Exciting! Let's get started!

## What is control flow?
Control flow describes the order in which individual statements, instructions, or function calls are executed or evaluated in a program. Here, let's look at one aspect of control flow: __conditional statements__.

* __Use case__: Suppose you want a block of code, for example, writing model results to a file, to execute only if a certain condition is met, such as the model converging. You can use an if-else statement to control whether that block of code runs. If the condition is `true`, the code runs; if `false`, it does not.
