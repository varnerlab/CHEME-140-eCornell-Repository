```julia
function texthashingvector(T::Array{String,1}, size::Int64)

    # initialize -
    x = Dict{Int,Int}();
    foreach(i-> x[i] = 0, range(0,step=1,length=size) |> collect)

    for s ∈ T
        h = hash(s);
        i = mod(h,size);
        x[i] += 1
    end

    # convert back to one-based array -
    result = zeros(size);
    for (k,v) ∈ x
        result[k+1] = v;
    end

    return result
end
```

```julia
example = ["I", "love", "NLP", "and", "I", "love", "programming", "in", "Julia","!"];
#example = ["Hello", "World", "!", "This", "is", "a", "test", "."];
```

```julia
v = texthashingvector(example, 10) |> println
```
