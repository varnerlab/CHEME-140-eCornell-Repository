## Text Representation Techniques
At the core of NLP is the need to convert text into a numerical representation that can be analyzed by machine learning algorithms. There are several techniques for doing this, but we'll focus on three fundamental concepts: __Tokenization, Hashing, and Bag-of-Words (BoW)__.

### Tokenization
Tokenization is the process of breaking down text into smaller units, called tokens. These tokens can be words, phrases, or even characters, depending on the level of granularity required for the analysis. Let's look at the formal definition of tokenization.

__Alphabet:__ Let's imagine our alphabet is represented by $\mathbf{\Sigma}$, e.g., Unicode code points. Then, a text string $s\in \mathbf{\Sigma}^*$, where $\mathbf{\Sigma}^*$ is the set of all finite-length sequences from the alphabet $\mathbf{\Sigma}$.
* _Uh, What?_ This is a little confusing, so let's break it down. The alphabet $\mathbf{\Sigma}$ is a set of objects that we can use to form _tokens_, e.g., subwords, words, sentences, etc. For example, our alphabet might include letters (a-z), numbers (0-9), punctuation marks (.,!?), and whitespace characters (spaces, tabs, newlines). Ok, then what is $\mathbf{\Sigma}^*$? 
* _Kleene Closure:_ Formally, $\mathbf{\Sigma}^{\star}$ is the Kleene Closure. It represents the set of all finite-length sequences (including the empty sequence $\epsilon$) that can be formed using the symbols from the alphabet $\mathbf{\Sigma}$. For example, if the alphabet is $\mathbf{\Sigma} = \{a, b\}$, then $\mathbf{\Sigma}^{\star} = \{\epsilon, \text{"a"}, \text{"b"}, \text{"ab"}, \text{"ba"}, \text{"aa"}, \text{"bb"}, \text{"aab"}, \ldots\}$. Thus, $\mathbf{\Sigma}^{\star}$ is all the tokens you can spell using your alphabet, of any finite length (including the empty word).

We've used the phrase _tokens_ a few times, but what exactly is a token? Let's define it formally.

__Tokens and Vocabulary__: Tokens are the individual units that result from a tokenization operation. Formally, a vocabulary  $\mathcal{V} \subseteq \mathbf{\Sigma}^{\star}$ is the set of all unique tokens that we want to consider in our analysis. 
* _Tokenizer function?_ The tokenizer function $\tau:\Sigma^{\star}\;\longrightarrow\;\mathcal V^{\star}$ maps a text string $s \in \mathbf{\Sigma}^*$ to a sequence of tokens $(t_1, t_2, \ldots, t_m) \in \mathcal{V}^*$, where $m$ is the number of tokens in the sequence, and $\mathcal{V}^*$ is the set of _all finite-length sequences_ of tokens from the vocabulary $\mathcal{V}$. 


Suppose, we have a text string $s = \text{"I love NLP!"}$. We can then tokenize $s$, using a space tokenizer function, into the vocabulary: $\mathcal{V} = \{\text{"I"}, \text{"love"}, \text{"NLP"}, \text{"!"}\}$. However, the set $\mathcal{V}^{\star}$ is a more general set given by:
$$
\mathcal{V}^{\star} = \{\,
  \underbrace{(t_1, t_2, \dots, t_n)}_{\text{any finite length }n\ge0}
  \mid
  t_i\in \mathcal{V}
\}.
$$
The $\mathcal{V}^{\star}$ set contains all finite-length sequences of tokens from the vocabulary $\mathcal{V}$, e.g., $\mathcal{V}^{\star} = \bigcup_{n=0}^{\infty} \mathcal{V}^n$, where $\mathcal{V}^n$ is the set of all sequences of length $n$ from the vocabulary $\mathcal{V}$. In general, each level $n$ has $|\mathcal{V}|^n$ elements, where $|\mathcal{V}|$ is the number of unique tokens in the vocabulary $\mathcal{V}$:


Below are example elements of $\mathcal{V}^n$ for $n=0,1,2,3$:

|  $n$  | Some elements of $\mathcal{V} ^n$                                                                                                                                                     |
| :---: | :------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **0** | $\varepsilon$ (the empty sequence)                                                                                                                                         |
| **1** | $(\text{"i"})$, $(\text{"!"})$, $(\langle\mathrm{unk}\rangle)$, …                                                                                                          |
| **2** | $(\text{"i"},\text{"i"})$, $(\text{"i"},\text{"love"})$, $(\text{"love"},\text{"nlp"})$, $(\text{"!"},\text{"!"})$, …                                                      |
| **3** | $(\text{"i"},\text{"love"},\text{"i"})$, $(\text{"nlp"},\text{"!"},\text{"nlp"})$, $(\langle\mathrm{unk}\rangle,\langle\mathrm{unk}\rangle,\langle\mathrm{unk}\rangle)$, … |

### Feature Hashing
Feature Hashing is a technique used to convert text into a fixed-size numerical representation. The key idea is to use a hash function to map tokens to a fixed-size vector, which can then be used for analysis. Let's look at particular feature algorithm called the _Weinberger-Moody Hashing Algorithm_.

```julia
function hashing(strings::Array{String,1}; 
    size::Int64 = 1024, hash::Dict{String,Int64} = nothing)::Array{Int64,1}

    # initialize -
    result = Array{Int64,1}(undef, size);
    fill!(result, 0); # initialize the result with 0s

    # iterate through the strings, and compute the hash
    frequency_dictionary = Dict{Int64, Int64}();
    for string ∈ strings
                
        h = hash[string]; # returns the position of the string in the corpus
        i = mod(h, size); # compute the index
        
        if (haskey(frequency_dictionary, i) == true)
            frequency_dictionary[i] += 1; # increment the count
        else
            frequency_dictionary[i] = 1; # initialize the count
        end
    end

    # populate the result array -
    kv = keys(frequency_dictionary) |> collect |> sort;
    for k ∈ kv
        result[k+1] = frequency_dictionary[k];
    end

    # return the result -
    return result;
end
```

```julia
v = let 
    
    v = Dict{String,Int}()
    testsentnce = "I love NLP!";
    tokens = split(testsentnce, " ") .|> String # tokenize the sentence by spaces
    for (i, token) ∈ enumerate(tokens)
        v[token] = i; # assign a unique index to each token
    end


    input = Array{String,1}(undef, 1);
    input[1] = "love";

    hashing(input; size=10, hash=v)
end
```

Let’s peel back the layers on

$$
V = \{\;\text{"i"},\;\text{"love"},\;\text{"nlp"},\;\text{"!"},\;\langle\mathrm{unk}\rangle\},
$$

and exactly what

$$
V^* = \bigcup_{n=0}^\infty V^n
$$

means.

---

## 1. By definition

1. **$V^0$**

   $$
     V^0 = \{\varepsilon\},
   $$

   the *singleton* set containing the *empty sequence*.

2. **$V^1$**

   $$
     V^1 = V
     = \{\,(\text{"i"}),\;(\text{"love"}),\;(\text{"nlp"}),\;(\text{"!"}),\;(\langle\mathrm{unk}\rangle)\}.
   $$

3. **$V^2$**
   All *ordered pairs* of tokens:

   $$
     V^2
     =
     \{\, (t_1, t_2) \mid t_1\in V,\;t_2\in V\}.
   $$

   There are $|V|^2 = 5^2 = 25$ of these, for example

   $$
     (\text{"i"},\text{"love"}),\;
     (\text{"nlp"},\text{"nlp"}),\;
     (\text{"!"},\langle\mathrm{unk}\rangle),\;\dots
   $$

4. **$V^3$, $V^4$, …**
   Similarly, $V^3$ has all triples $(t_1,t_2,t_3)$, $V^4$ all 4-tuples, and so on.

5. **Finally**

   $$
     V^* = V^0 \;\cup\; V^1 \;\cup\; V^2 \;\cup\; V^3 \;\cup\cdots.
   $$

   Any finite sequence of tokens of *any* length—including length zero—is in $V^*$.

---

## 2. Concrete examples

Below are *all* the elements of $V^n$ for $n=0,1,2,3$ (a few select ones for $n=2,3$):

|  $n$  | Some elements of $V^n$                                                                                                                                                     |
| :---: | :------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **0** | $\varepsilon$ (the empty sequence)                                                                                                                                         |
| **1** | $(\text{"i"})$, $(\text{"!"})$, $(\langle\mathrm{unk}\rangle)$, …                                                                                                          |
| **2** | $(\text{"i"},\text{"i"})$, $(\text{"i"},\text{"love"})$, $(\text{"love"},\text{"nlp"})$, $(\text{"!"},\text{"!"})$, …                                                      |
| **3** | $(\text{"i"},\text{"love"},\text{"i"})$, $(\text{"nlp"},\text{"!"},\text{"nlp"})$, $(\langle\mathrm{unk}\rangle,\langle\mathrm{unk}\rangle,\langle\mathrm{unk}\rangle)$, … |

Any of these is a valid element of $V^*$.

---

## 3. Intuition

* **Think “sentences”**:
  If your tokens are words and punctuation, then $V^*$ is the set of *all possible sentences* you can type (including the empty sentence).

* **Length grows**:

  * $V^0$ has just 1 element (the empty).
  * $V^1$ has 5 elements.
  * $V^2$ has $5^2=25$.
  * $V^3$ has $5^3=125$.
  * …and so on.

* **No upper bound**:
  You can make arbitrarily long sequences, so $V^*$ is infinite, even though $V$ itself is finite.

---

### Key takeaway

$$
V^* = \{\,
  \underbrace{(t_1, t_2, \dots, t_n)}_{\text{any finite length }n\ge0}
  \mid
  t_i\in V
\}.
$$

It’s simply *every* finite list of tokens drawn from your vocabulary.


Here’s the same formalization with corrected LaTeX and a bit more precision in the notation:

---

## 1. Basic setup

* **Alphabet**
  Let $\Sigma$ be a finite set of characters (e.g.\ Unicode code‐points).
  A **text string** is any element \$T\in\Sigma^\*\$ (the set of all finite sequences over \$\Sigma\$).

* **Vocabulary**
  Let \$\mathcal V\$ be a finite set of tokens (words, subwords, punctuation, etc.).

* **Token IDs**
  We assign each token an integer via
  $\mathrm{id}:\mathcal V\to\{1,2,\dots,|\mathcal V|\}\,. $

---

## 2. The tokenizer function

Define the **tokenizer** as a map

$$
\tau:\Sigma^*\;\longrightarrow\;\mathcal V^*, 
\qquad
\tau(T) = (t_1,\dots,t_m)\,,
$$

together with an inverse‐mapping \$\text{str}:\mathcal V\to\Sigma^\*\$ such that

$$
T = \text{str}(t_1)\;\bigl\|\;\text{str}(t_2)\;\bigl\|\;\cdots\;\bigl\|\;\text{str}(t_m),
$$

where \$|\$ denotes string concatenation.

---

## 3. Examples

### 3.1 Whitespace‐based (word) tokenization

Split \$T\$ at runs of whitespace into maximal non‐empty substrings.  Equivalently, choose indices

$$
0 = i_0 < i_1 < \cdots < i_m = |T|
$$

such that for each \$k\$

1. the substring \$T\[i\_{k-1}+1 ,\dots, i\_k]\$ contains no whitespace, and
2. either \$i\_k=|T|\$ or \$T\[i\_k+1]\$ is whitespace.

Then

$$
\tau_{\rm ws}(T) \;=\;(t_1,\dots,t_m),
\qquad
t_k \;=\;T\bigl[i_{k-1}+1\,\dots\,i_k\bigr].
$$

### 3.2 Greedy subword (BPE/WordPiece)

1. **Vocabulary construction** via \$K\$ merges:

   * Start with \$\mathcal V\_0=\$ all characters.
   * For \$k=1,\dots,K\$, pick the most frequent adjacent pair \$(a,b)\$ in your corpus, merge it into a new symbol \$c\$, and set
     $\mathcal V_k \;=\;\mathcal V_{k-1}\,\cup\{c\}.$
   * Finally \$\mathcal V = \mathcal V\_K\$.

2. **Segmentation** by longest‐match:
   Define

   $$
   \tau_{\rm BPE}(T) =
   \begin{cases}
   () , & T=\epsilon,\\
   (v)\,\bigl\|\;\tau_{\rm BPE}(T[|v|+1\,\dots\,|T|]) , &\text{otherwise},
   \end{cases}
   $$

   where

   $$
   v = \arg\max\bigl\{\,|u|\;:\;u\in\mathcal V,\;u\text{ is a prefix of }T\bigr\}.
   $$

---

## 4. From text to model input

The full pipeline is

$$
T \;\xrightarrow{\;\tau\;}\;(t_1,\dots,t_m)
\;\xrightarrow{\;\mathrm{id}\;}\;(i_1,\dots,i_m),
$$

where \$(i\_1,\dots,i\_m)\$ is the integer sequence fed into your model.

---

**Key point:** tokenization is just a particular choice of finite vocabulary \$\mathcal V\$ plus a segmentation map \$\tau:\Sigma^*\to\mathcal V^*\$. Different schemes (words, subwords, characters) all fit in this same formal framework.
