# Read: Simple Sorting Algorithms for Arrays
One of the common tasks in programming is sorting data. Sorting an array orders its elements (e.g., ascending or descending) so that you can perform efficient searches, aggregations, and other algorithms that require or benefit from organized data.

Let's take a look at some simple sorting algorithms for arrays:
* __Bubble sort__ repeatedly iterates through an array, comparing adjacent elements, and swaps them if they are in the wrong order. The worst-case time complexity of Bubble sort is $\mathcal{O}(n^2)$, where $n$ is the number of elements in the unsorted list.
* __Insertion sort__ iterates through the list and compares each element with the elements before it, and if necessary, it performs shift operations to sort a random list into ascending order. It has a worst-case time complexity of $\mathcal{O}(n^2)$ when the list is in reverse order.

## Bubble Sort
Bubble sort is a simple method of ordering a list. It involves repeatedly passing through the list, comparing adjacent items, and swapping any two neighboring items that are out of order. The process is repeated until the list is sorted.

Let's look at some pseudo-code for Bubble sort:

__Initialization:__ The algorithm starts with an unsorted list of elements in the vector $\mathbf{a}$. Compute the length of the unsorted list $n \gets \texttt{length}(a)$. Set the flag $\text{swapped} \gets \texttt{true}$ to indicate that a swap has occurred.

While $\text{swapped}$ is $\texttt{true}$ __do__:
1. Set a flag $\text{swapped} \gets \texttt{false}$.
2. For each $i\in\left\{1,\ldots,n-1\right\}$ __do__:
    * If $a_{i} > a_{i+1}$ __then__:
        * Call the $\texttt{swap}(a_{i}, a_{i+1})$ method to swap (exchange) the two elements.
        * Set the flag $\text{swapped} \gets \texttt{true}$ (indicating a swap was made)
3. Update the length of the list $n \gets n - 1$ (reduce the length of the list by 1). Go to the while statement.
4. If $\text{swapped} \gets \texttt{false}$ then: the vector $\mathbf{a}$ is sorted, and the algorithm terminates.

where the $\texttt{swap}(a_{i}, a_{i+1})$ method does the following:

__Initalize__: Given two elements $a_{i}$ and $a_{i+1}$, we will swap them if they are out of order. The algorithm starts with the first element $a_{i}$ and the second element $a_{i+1}$.
1. Set a temporary variable $t \gets a_{i}$.
2. Set $a_{i} \gets a_{i+1}$.
3. Set $a_{i+1} \gets t$.
4. Return the swapped elements $a_{i}$ and $a_{i+1}$

### Worst-case time complexity
The worst-case time complexity of Bubble sort is $\mathcal{O}(n^2)$, where $n$ is the number of elements in the unsorted list. Let's see where this comes performance bound comes from.

#### Analysis
The number of comparisons is $n-1$ for the first pass, $n-2$ for the second pass, and so on, until we reach 1 comparison. This gives us a total of $n-1 + n-2 + \ldots + 1 = \frac{(n-1)(n)}{2}$ comparisons. Further, in the worst case, the number of swaps is at most the number of comparisons, i.e., we make a swap for each comparison. Thus, we expect $\frac{(n-1)(n)}{2}$ swaps. Let the time per comparison be $c$ and the time per swap be $s$ (constant). Then, the total time as a function of $n$, denoted as $T(n)$ is given by: 
$$\begin{align*}
T(n) &= \underbrace{c\cdot\left(\frac{(n-1)(n)}{2}\right)}_{\text{comparisons}} + \underbrace{s\cdot\left(\frac{(n-1)(n)}{2}\right)}_{\text{swaps}} \\
&= c\cdot\frac{n^2 - n}{2} + s\cdot\frac{n^2 - n}{2} \\
&= \left(c + s\right)\cdot\frac{n^2 - n}{2} \\
&\sim \mathcal{O}(n^2)\,\blacksquare
\end{align*}$$

## Insertion Sort
Insertion sort works like arranging playing cards in your hand: you take each new item and slide it into its correct position among the ones you’ve already sorted. Unlike bubble sort—which repeatedly swaps neighboring items until everything is in order—insertion sort builds a sorted section of the list one element at a time by inserting each element where it belongs. Let's look at some pseudo-code for Insertion sort:

__Initialization:__ The algorithm starts with an unsorted list of elements in the vector $\mathbf{a}$. Compute the length of the unsorted list $n \gets \texttt{length}(a)$.

For each $j\in\left\{2,\ldots,n\right\}$ __do__:
1. Set a temporary variable $\text{key} \gets a_{j}$.
2. Set $i \gets j - 1$.
3. While $i \geq 1$ and $a_{i} > \text{key}$ __do__:
    * Set $a_{i+1} \gets a_{i}$.
    * Set $i \gets i - 1$.
4. Set $a_{i+1} \gets \text{key}$.

### Worst-case time complexity
The worst-case time complexity of insertion sort is also $\mathcal{O}(n^2)$, where $n$ is the number of elements in the unsorted list. Let's see where this performance bound comes from.

#### Analysis
Let the time per comparison be $c$ and the time per shift be $s$ (constant). The number of comparisons in the worst case is $n-1 + n-2 + \ldots + 1 = \frac{(n-1)(n)}{2}$, and the number of shifts is also at most $\frac{(n-1)(n)}{2}$. Thus, we expect $\frac{(n-1)(n)}{2}$ comparisons and shifts. The total time as a function of $n$, denoted as $T(n)$ is given by:
$$\begin{align*}
T(n) &= \underbrace{c\cdot\left(\frac{(n-1)(n)}{2}\right)}_{\text{comparisons}} + \underbrace{s\cdot\left(\frac{(n-1)(n)}{2}\right)}_{\text{shifts}} \\
&= c\cdot\frac{n^2 - n}{2} + s\cdot\frac{n^2 - n}{2} \\
&= \left(c + s\right)\cdot\frac{n^2 - n}{2} \\
&\sim \mathcal{O}(n^2)\,\blacksquare
\end{align*}$$
which is exactly the same as the worst-case time complexity of Bubble sort!

### Caution
Both Bubble sort and Insertion sort have a worst-case time complexity of $\mathcal{O}(n^2)$, making them inefficient for large datasets. They are mainly used for educational purposes or small datasets where their simplicity is beneficial. More efficient algorithms like Quick Sort or Merge Sort are preferred for larger datasets due to their superior time complexities.
* __Quick Sort__ is a _divide-and-conquer_ sorting algorithm that selects a _pivot_ element from the array and partitions the other elements into two sub-arrays according to whether they are less than or greater than the pivot. The sub-arrays are then sorted _recursively_. Its average-case time complexity is $\mathcal{O}(n\log n)$, but its worst-case time complexity can degrade to $\mathcal{O}(n^2)$ if the pivot is not chosen wisely.
* __Merge Sort__ is a _divide-and-conquer_ sorting algorithm that _recursively_ splits an input array into two halves, sorts each half, and then merges the two sorted halves into a single sorted array. Because each split reduces the problem size by half and the merge step takes linear time, its worst‐case and average‐case time complexity is $\mathcal{O}(n\log n)$, where $n$ is the number of elements in the array. 

Quick Sort and Merge Sort are more efficient than Bubble Sort and Insertion Sort for large datasets, with average-case time complexities of $\mathcal{O}(n\log n)$. Both are __recursive__, breaking the problem into smaller sub-problems.
* _What is recursion?_ Recursion is a programming technique where a function calls itself to solve smaller instances of the same problem. This idea is used in Quick Sort and Merge Sort to break down complex problems into simpler ones. We'll see this in more detail later in the course.
