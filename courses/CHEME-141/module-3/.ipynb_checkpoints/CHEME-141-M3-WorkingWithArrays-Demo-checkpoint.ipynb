# Demo: Working with Arrays
Arrays are a fundamental _linear data structure_ used in many (if not most) programming languages, e.g., Julia, Python, or C. Let's start with some fun facts and key concepts about arrays:

* _What is an array?_ An array is a fixed-size, _homogeneous_ collection of elements—each of the same type—stored in a contiguous memory block. This contiguous layout minimizes per-element overhead and provides constant-time access by index.
* _Constant time access?_ Constant-time access means that accessing or modifying an element at any given index in the array takes a fixed, bounded amount of time, which we represent as $\mathcal{O}(1)$—regardless of the array’s length. This is achieved by storing elements in contiguous memory and using simple arithmetic on the index to directly compute the element’s address.
* _What is big $\mathcal{O}$ notation?_ Big $\mathcal{O}$ is a mathematical notation used to describe the upper bound of an algorithm's time (or space) _complexity_. It provides a high-level understanding of how the performance of an operation (or entire algorithm) scales with input size, focusing only on the most significant factors while ignoring constant factors and lower-order terms.

Let's dive into some code examples to see how arrays work.  In this demo, we will explore how to create and manipulate arrays in Julia's [built-in array type](https://docs.julialang.org/en/v1/base/arrays/#Core.Array-Tuple%7BNothing,%20Any%7D). Let's go!
