## Example 1: Build a Stack holding a molecular formula string
A [stack](https://en.wikipedia.org/wiki/Stack_(abstract_data_type)) is a linear data structure that follows the __Last-In-First-Out (LIFO)__ principle, where elements are added and removed from the __top of the stack__. It provides constant-time operations for pushing an element onto the stack and popping it from the stack, i.e., accessing the top element.

To explore these concepts, let's specify a compound string encoded in the `compound_for_stack::String` variable to load onto the stack.

```julia
compound_for_stack = "C6H12O6"; # Glucose
```

### Populate

Next, we'll create a stack holding the characters of the `compound_for_stack::String` variable. Let's step through what is happening in the code block below:
* First, we initialize an empty stack that will hold the characters of the `compound_for_stack::String` variable. Why characters? Remember that String types can be rewritten as arrays of characters, so we'll do the same, but with a stack.
* Next, we use [the `collect(...)` method](https://docs.julialang.org/en/v1/base/collections/#Base.collect-Tuple%7BAny%7D) which takes its argument and converts that to an array, in this case the `chars::Array{Char,1}` variable.
* Finally, we directly iterate through the characters in the `chars::Array{Char,1}` variable using a $\texttt{for}$ c $\in$ chars construct and push each character $c$ onto the stack $s$. We then return the populated stack to the caller in the `compound_stack::Stack{Char}` variable.

```julia
compound_stack = let 
    
    # initialize -
    s = Stack{Char}();

    # add string to the stack -
    chars = collect(compound_for_stack); # convert string to char array
    for c ∈ chars
        push!(s,c);
    end

    s; # return the
end
```

### Access
We access every element in a [stack](https://en.wikipedia.org/wiki/Stack_(abstract_data_type)) using [a while-loop](https://docs.julialang.org/en/v1/base/base/#while). In this pattern, we continue looping until the `compound_stack::Stack{Char}` is empty. 

We use [the `isempty(...)` method](https://docs.julialang.org/en/v1/base/collections/#Base.isempty) to check if the `compound_stack::Stack{Char}` is empty. This method takes a collection and returns `true` if it is empty and `false` otherwise.

```julia
while (isempty(compound_stack) == false)
    c = pop!(compound_stack)
    @show (c, compound_stack)
end
```

When we access the top element of the stack, using [the `pop!()` method](https://juliacollections.github.io/DataStructures.jl/stable/stack_and_queue/) we get the last element that was added to the stack. This follows the _Last-in-First-Out (LIFO)_ principle.

## Example 2: Build a Queue holding a molecular formula string
A [queue](https://en.wikipedia.org/wiki/Queue_(abstract_data_type)) is a linear data structure that operates on the __First-In-First-Out (FIFO)__ principle, where the first element added to the queue is the first one to be removed. It allows for efficient operations for adding elements to the rear of the queue and removing elements from the front, making it ideal for scenarios like task scheduling and resource management.

To explore these concepts, let's specify a compound string encoded in the `compound_for_queue::String` variable to load onto the [queue](https://en.wikipedia.org/wiki/Queue_(abstract_data_type)).

```julia
compound_for_queue = "C10H16N5P13P3"; # ATP
```

### Populate
Next, we'll create a stack holding the characters of the `compound_for_queue::String` variable. Let's step through what is happening in the code block below:
* First, we initialize an empty stack that will hold the characters of the `compound_for_queue::String` variable. Why characters? Remember that String types can be rewritten as arrays of characters, so we'll do the same, but with a queue.
* Next, we use [the `collect(...)` method](https://docs.julialang.org/en/v1/base/collections/#Base.collect-Tuple%7BAny%7D) which takes its argument and converts that to an array, in this case the `chars::Array{Char,1}` variable.
* Finally, we directly iterate through the characters in the `chars::Array{Char,1}` variable using a $\texttt{for}$ c $\in$ chars construct and push each character $c$ onto the queue $s$. We then return the populated queue to the caller in the `compound_queue::Queue{Char}` variable

```julia
compound_queue = let 
    
    # initialize -
    q = Queue{Char}();

    # add string to the queue -
    chars = collect(compound_for_queue); # convert string to char array
    for c ∈ chars
        enqueue!(q,c);
    end

    q; # return the
end
```

### Access

We access elements from the queue using [the `pop!(...)` method](https://juliacollections.github.io/DataStructures.jl/stable/stack_and_queue/). The [`pop!(...)` method](https://juliacollections.github.io/DataStructures.jl/stable/stack_and_queue/) removes the top element from the queue (returns it to you) and then resets the top element.
* We can access every element in a [queue](https://en.wikipedia.org/wiki/Queue_(abstract_data_type)) using [a `while-loop`](https://docs.julialang.org/en/v1/base/base/#while) in combination [with the `dequeue!(...)` method](https://juliacollections.github.io/DataStructures.jl/stable/stack_and_queue/). We continue looping until the `compound_queue::Queue{Char}` is empty, where we use [the `isempty(...)` method](https://docs.julialang.org/en/v1/base/collections/#Base.isempty) to check if `compound_stack::Queue{Char}` is empty. The [`isempty(...)` method](https://docs.julialang.org/en/v1/base/collections/#Base.isempty) takes a collection and returns `true` if it is empty and `false` otherwise.

```julia
while (isempty(compound_queue) == false)
    c = dequeue!(compound_queue) # notice the ! = mutating operation
    @show (c, compound_queue)
end
```

When we access the elements in the queue using [the `dequeue!(...)` method](https://juliacollections.github.io/DataStructures.jl/stable/stack_and_queue/), we access them in the order they were added to the queue. The first element added to the queue is the first one to be removed, and so on. This is known as the _First-in-First-Out (FIFO)_ principle.
