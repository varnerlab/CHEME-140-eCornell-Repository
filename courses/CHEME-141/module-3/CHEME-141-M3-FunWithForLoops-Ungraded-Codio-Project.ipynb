## Example 1: Basic for-loop iteration example for an Array
Let's start by looking at the basic structure of [a `for-loop.`](https://docs.julialang.org/en/v1/base/base/#for) A [`for-loop`](https://docs.julialang.org/en/v1/base/base/#for) has a header that specifies how many times the loop will iterate. The loop `index` (the `iteration variable`) is passed into the loop's body, where you put your logic. The `iteration variable` is always a new variable, even if a variable of the same name exists in the enclosing scope. 
* In Julia, the [`for-loop`](https://docs.julialang.org/en/v1/base/base/#for) has its own local scope that captures variables from the outside but doesn't pass new variables created inside the loop to the outside unless they already exist. The [local scope of the for loop](https://docs.julialang.org/en/v1/manual/variables-and-scoping/#local-scope) ends with the `end` keyword.
* We use the [println function](https://docs.julialang.org/en/v1/base/io-network/#Base.println) to show output from the loop. This function takes a [`String`](https://docs.julialang.org/en/v1/manual/strings/) as an argument and sends it to `stdout` (the default output destination) or to an output specified by the caller. The `$(...)` is an example of a [String interpolation](https://docs.julialang.org/en/v1/manual/strings/#string-interpolation) operation.

```julia
number_of_elements = 5;
random_vector_array = rand(number_of_elements); # create an array of random Floating point values
value = nothing
for i in 1:number_of_elements
    value = random_vector_array[i];
    println("The index i = $(i) and the value = $(random_vector_array[i])");
end # end of for loop scope
value
```

Another [`for-loop`](https://docs.julialang.org/en/v1/base/base/#for) pattern is the [eachindex pattern](https://docs.julialang.org/en/v1/base/arrays/#Base.eachindex). We use the [eachindex pattern](https://docs.julialang.org/en/v1/base/arrays/#Base.eachindex) when we don't explicitly know how many elements we have in an ordered collection, but we want to visit all of them in order: 

The [eachindex pattern](https://docs.julialang.org/en/v1/base/arrays/#Base.eachindex) is the preferred pattern compared with something like `for i in 1:length(random_vector_array)` for several important reasons. 
* First, `eachindex` works correctly with specialized array types, including those with custom indexing (available through packages like `OffsetArrays.jl`), while `1:length(array)` assumes standard one-based indexing. 
* Second, `eachindex` is often more efficient because it can optimize the iteration based on the specific array type and memory layout. 
* Finally, `eachindex` makes your code more generic and robust, working seamlessly with different array types and dimensions without requiring you to know the specific indexing scheme.

Let's see an example of the [eachindex pattern](https://docs.julialang.org/en/v1/base/arrays/#Base.eachindex) in action:

```julia
# N = length(random_vector_array)
for i ∈ eachindex(random_vector_array)
    value = random_vector_array[i];
    println("The index i = $(i) and the value = $(value)");
end
```

If we don't care about the element index $i$ in a collection but instead want the values, we can iterate over the elements directly. For example, the code block below accesses the values of the `random_vector_array`  directly but `NOT` their indexes:

```julia
for value ∈ random_vector_array
    println("The value = $(value)");
end
```

Finally, we can iterate over other types of ordered collections, e.g., [tuples](https://docs.julialang.org/en/v1/manual/functions/#Tuples), which are fixed-length ordered containers that can hold any values, but cannot be modified once constructed, i.e., they are immutable. Let's build a `tuple` holding some `Int` types and iterate over these using the [eachindex pattern](https://docs.julialang.org/en/v1/base/arrays/#Base.eachindex).
* The previous examples used the [println function](https://docs.julialang.org/en/v1/base/io-network/#Base.println) to print output; here, we show another approach that does the same thing, namely the [@show macro](https://docs.julialang.org/en/v1/base/base/#Base.@show) which prints one or more expressions, and their results, to `stdout`, and returns the last result:

```julia
example_tuple = (1,2,3,6,5,4,1,1,1,1) # tuple, not an array!
for i ∈ eachindex(example_tuple)
    @show (i, example_tuple[i])
end
```

## Example 2: Basic for-loop iteration example for an unordered Set
One of the central differences between linear and nonlinear data structures is the organization of the data and how we traverse the items in the data structure. In the case of unordered non-linear collections such as [Sets](https://docs.julialang.org/en/v1/base/collections/#Base.Set), we can use a [`for-loop`](https://docs.julialang.org/en/v1/base/base/#for) to visit each element of the collection. However, because the original collection is unordered, the order in which we visit the elements will be random. 

In the specific case of [Sets](https://docs.julialang.org/en/v1/base/collections/#Base.Set), which model `bags of stuff`, there is no notion of index (which can be a little confusing). Thus, we can only access the items directly. The other exciting thing about [Sets](https://docs.julialang.org/en/v1/base/collections/#Base.Set) is that they are `unique`, meaning there are no repeated elements. Notice in the example below that we include the character 'A' twice in our input array, but it appears only once when we iterate over the Set because Sets automatically eliminate duplicates.

```julia
example_set = Set{Char}(['A','B','C','D','R','U','S','T','A']);
for value ∈ example_set
    @show value
end
```

## Example 3: Basic for-loop iteration of a Dictionary
Dictionaries are also examples of non-linear data structures, which do not sequentially arrange data. Instead, data can be connected in a hierarchical or network-based format, allowing for more complex relationships between the elements. However, we can still traverse a dictionary [using a `for-loop`](https://docs.julialang.org/en/v1/base/base/#for).

For this example, we'll use real financial data to demonstrate dictionary iteration in a practical context. We gathered a daily open-high-low-close `dataset` for each firm in the [S&P500](https://en.wikipedia.org/wiki/S%26P_500) from `01-03-2014` until `12-29-2024`, along with data for a few exchange-traded funds and volatility products during that time. 

Let's load the `original_dataset` by calling [the `MyTrainingMarketDataSet` method](https://varnerlab.github.io/VLDataScienceMachineLearningPackage.jl/dev/data/#VLDataScienceMachineLearningPackage.MyTrainingMarketDataSet) (which takes no arguments) and remove firms that do not have the maximum number of trading days. The cleaned dataset $\mathcal{D}$ is stored in the `dataset` variable, where the dataset $\mathcal{D}$ has data for $\mathcal{L}$ firms. When we use a [`for-loop`](https://docs.julialang.org/en/v1/base/base/#for) with a [Dictionary](https://docs.julialang.org/en/v1/base/collections/#Dictionaries), which is an unordered collection of `key=>value` pairs, we get both the `key` and `value` as the iteration variable organized as a tuple.

```julia
original_dataset = MyTrainingMarketDataSet() |> x-> x["dataset"]
```

The `original_dataset::Dict{String,DataFrame}` is a dictionary where the `key` is the firm ticker symbol and the `value` is a [DataFrame](https://juliadata.github.io/DataFrames.jl/stable/) containing the daily open-high-low-close (and some other) data for that firm. 

Let's check out an example firm: `AAPL`

```julia
original_dataset["AAPL"] # We access the AAPL data frame by passing the key "AAPL" to the dataset dictionary
```

### Clean the data
Not all tickers in our dataset have the maximum number of trading days for various reasons, e.g., acquisition or de-listing events. Let's collect only those tickers with the maximum number of trading days.

First, let's compute the number of records for a company that we know has a maximum value, e.g., `AAPL`, and save that value in the `maximum_number_trading_days` variable:

```julia
maximum_number_trading_days = original_dataset["AAPL"] |> nrow; # nrow? (check out: DataFrames.jl)
```

Now, let's iterate through our data and collect only tickers with `maximum_number_trading_days` records. Save that data in the `dataset::Dict{String,DataFrame}` variable:

```julia
dataset = Dict{String, DataFrame}();
for (ticker,data) ∈ original_dataset    
    if (nrow(data) == maximum_number_trading_days)
        dataset[ticker] = data;
    else
        println("Ticker $(ticker) does not have full data!")
    end
end
dataset;
```

The cleaned dataset should have the same number or fewer elements than the original dataset. Let's verify this by checking the final count:

```julia
dataset |> length # pipe operator: passes dataset as first argument to length function
```

## Summary: Understanding For Loop Patterns

Through these three examples, we've explored how for loops adapt to different data structures while maintaining consistent syntax. With arrays and tuples, we can choose between index-based iteration (when we need position information) or direct value iteration (when we only care about the data). The `eachindex` pattern provides a robust, efficient way to iterate when we need indices but want our code to work with any array type.

Sets demonstrate how for loops handle unordered collections, where iteration order is unpredictable but the syntax remains familiar. The automatic elimination of duplicates makes Sets particularly useful for collecting unique items from larger datasets.

Dictionaries showcase the power of tuple unpacking in for loops, allowing us to simultaneously access both keys and values. This pattern is especially valuable for data processing tasks, as we saw with the financial dataset cleaning example.

The key insight is that Julia's for loop syntax remains consistent across different data structures, but the iteration behavior adapts to respect each structure's inherent properties. This consistency makes for loops a versatile tool for data analysis and scientific computing, regardless of how your data is organized.

### Beyond For Loops

While for loops are fundamental to understanding iteration in Julia (and all other programming languages), the Julia language offers additional powerful iteration patterns worth exploring as we advance through the course:
* The [`foreach()` function](https://docs.julialang.org/en/v1/base/collections/#Base.foreach) provides a functional programming approach to iteration, particularly useful when you want to apply a function to each element without creating a new collection. 
* [Array comprehensions](https://docs.julialang.org/en/v1/manual/arrays/#man-comprehensions) offer a concise way to create new arrays by transforming existing data, combining iteration and transformation in a single expression. 

These patterns complement for loops and can often lead to more expressive and efficient code for specific use cases.
